# 4-6. Implicit Differentiation


### Problem 1. Power Law
From $pv^a=C$, logarithmic differentiation gives $dp/p+a\,dv/v=0$. Hence
$$\boxed{dv/dp=-v/(ap).}$$
Differentiating $v=Kp^{-1/a}$ gives
$$\boxed{d^2v/dp^2=\frac{a+1}{a^2}\frac{v}{p^2}.}$$


### Problem 2. Equation $ye^{xy}=\sin x$
Differentiate once:
$$
y'e^{xy}+ye^{xy}(y+xy')=\cos x.
$$
At $(0,0)$ this gives $y'=1$. Differentiating the series form $y e^{xy}=y+xy^2+\cdots=\sin x=x-x^3/6+\cdots$ gives $y=x+O(x^3)$, so
$$\boxed{y'(0)=1,\qquad y''(0)=0.}$$


### Problem 3. Equation $x^y=y^x$
Taking logs gives $y\ln x=x\ln y$. Differentiate:
$$
y'\ln x+\frac{y}{x}=\ln y+x\frac{y'}{y}.
$$
Thus
$$
\boxed{y'=\frac{\ln y-y/x}{\ln x-x/y}.}
$$
At $(2,4)$,
$$\boxed{y'=\frac{2\ln2-2}{\ln2-1/2}.}$$


### Problem 4. Equation $xe^y=ye^x$
Taking logs gives $\ln x+y=\ln y+x$. Thus
$$
\frac1x+y'=\frac{y'}y+1,
\qquad
\boxed{y'=\frac{1-1/x}{1-1/y}.}
$$
A second differentiation of this implicit relation gives the requested $y''$; substituting the displayed $y'$ is the cleanest final form.


### Problem 5. Tangent Line for $xy^3-yx^3=6$
Differentiate:
$$
y^3+3xy^2y'-y'x^3-3x^2y=0.
$$
At $(1,2)$, $8+12y'-y'-6=0$, so $y'=-2/11$. Therefore
$$\boxed{y-2=-\frac2{11}(x-1).}$$


### Problem 6. Second Derivative for Problem 5
Differentiate the first-derivative equation again and substitute $(x,y,y')=(1,2,-2/11)$. Solving for $y''$ gives
$$\boxed{y''=-\frac{1320}{1331}.}$$


### Problem 7. Tangent Line for $y^3-x^2y=8$
Differentiate:
$$
3y^2y'-2xy-x^2y'=0,
\qquad
 y'=\frac{2xy}{3y^2-x^2}.
$$
At $(3,-1)$, $y'=1$, so
$$\boxed{y+1=x-3.}$$


### Problem 8. Second Derivative for Problem 7
Differentiate $3y^2y'-2xy-x^2y'=0$ again and substitute $(3,-1,y')=(3,-1,1)$. This gives
$$\boxed{y''=-\frac32.}$$


### Problem 9. Astroid Tangents
For $x^{2/3}+y^{2/3}=4$,
$$
\frac{2}{3}x^{-1/3}+\frac{2}{3}y^{-1/3}y'=0,
\qquad
\boxed{y'=-\left(\frac{y}{x}\right)^{1/3}.}
$$
At $(2\sqrt2,-2\sqrt2)$, $y'=1$. At $(8,0)$ the tangent is vertical, $\boxed{x=8}$. At $(0,8)$ the tangent is horizontal, $\boxed{y=8}$.


### Problem 10. Tangent at the Origin
For $xe^y+ye^x=0$,
$$
e^y+xe^y y'+y'e^x+ye^x=0.
$$
At $(0,0)$, $1+y'=0$, so the tangent is
$$\boxed{y=-x.}$$


### Problem 11. Second Derivative at the Origin
Differentiating the first-derivative equation again and substituting $x=y=0$, $y'=-1$ gives
$$\boxed{y''=0.}$$
