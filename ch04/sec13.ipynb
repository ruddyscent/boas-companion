# 4-13. Miscellaneous Problems


### Problem 1. Euler's Theorem for Homogeneous Functions
Differentiate $f(tx,ty,tz)=t^nf(x,y,z)$ with respect to $t$:
$$xf_1(tx,ty,tz)+yf_2(tx,ty,tz)+zf_3(tx,ty,tz)=nt^{n-1}f(x,y,z).$$
Set $t=1$ to get
$$\boxed{x f_x+y f_y+z f_z=nf.}$$


### Problem 2. Distance from a Point to a Line
The distance from $(x_0,y_0)$ to $ax+by=c$ is
$$\boxed{D=\frac{|ax_0+by_0-c|}{\sqrt{a^2+b^2}}.}$$
For $(2,1)$ and $3x+2y=4$, $D=|6+2-4|/\sqrt{13}=\boxed{4/\sqrt{13}}$.


### Problem 3. Mixed Partial of $y$
With $y=r\sin\theta$ and $r=x\sec\theta$ when $x,\theta$ are variables, write $y=x\tan\theta$. Then
$$\boxed{\partial^2y/(\partial x\partial\theta)=\sec^2\theta.}$$


### Problem 4. Mixed Partial of $r$
With $r=y\csc\theta$,
$$r_y=\csc\theta,
\qquad \boxed{r_{\theta y}=-\csc\theta\cot\theta.}$$


### Problem 5. Coordinate-Dependent Partials
For $z=y^2-2x^2$:
$$\boxed{(z_x)_r=-2x-2x= -4x,}$$
using $y^2=r^2-x^2$.
With $x$ fixed, $y=x\tan\theta$, so $z=x^2\tan^2\theta-2x^2$ and
$$\boxed{(z_\theta)_x=2x^2\tan\theta\sec^2\theta.}$$
Then differentiate this with respect to $x$ for the mixed partial.


### Problem 6. Partials for $z=r^2-x^2$
Since $z=r^2-r^2\cos^2\theta=r^2\sin^2\theta$,
$$\boxed{(z_r)_\theta=2r\sin^2\theta,
\quad (z_\theta)_r=2r^2\sin\theta\cos\theta,
\quad z_{r\theta}=4r\sin\theta\cos\theta.}$$
In rectangular form, $z=y^2$, so $\boxed{(z_x)_y=0}$.


### Problem 7. Relative Error
For $f=x^3y^2$,
$$df/f=3dx/x+2dy/y.$$
Two $1\%$ errors give about $\boxed{5\%}$ worst case.


### Problem 8. Rope Around a Sphere
If circumference increases by $2$ ft, then $2\pi dh=2$, so
$$\boxed{dh=1/\pi\text{ ft}\approx3.8\text{ in}.}$$
The answer is independent of the planet's radius, so it is the same for Earth and Moon.


### Problem 9. Derivative from Two Equations
Differentiate the two equations to solve for $dx/dt$ and $dy/dt$, then use
$$\boxed{dz/dt=yx^{y-1}x'+x^y\ln x\,y'}$$
for $z=x^y$.


### Problem 10. Power in Polar Variables
For $w=(r\cos\theta)^{r\sin\theta}$,
$$\ln w=r\sin\theta\ln(r\cos\theta).$$
Thus
$$\boxed{w_\theta=w\,r\{\cos\theta\ln(r\cos\theta)-\sin\theta\tan\theta\}.}$$


### Problem 11. Ellipse Derivatives
From $x^2/a^2+y^2/b^2=1$,
$$\boxed{y'=-\frac{b^2x}{a^2y}.}$$
Differentiating again gives
$$\boxed{y''=-\frac{b^2}{a^2y}-\frac{b^4x^2}{a^4y^3}.}$$


### Problem 12. Partial at a Point
Use $dz=(2r+st)dr+(2s+rt)ds+rsdt$ and differentiate the constraint with $dt=0$. At $r=2$, $s=t=1$, solve for $ds/dr$ and substitute into $dz/dr$.


### Problem 13. Partial from a Two-Equation System
Differentiate the two equations, set $dy=0$, substitute the point, and solve the resulting two linear equations for $ds/dt$. This is the requested $(\partial s/\partial t)_y$.


### Problem 14. Composite Partial
For $w=f(x,s,t)$, $s=2x+y$, $t=2x-y$, at fixed $y$:
$$\boxed{(w_x)_y=f_x+2f_s+2f_t.}$$


### Problem 15. Three-Argument Composite
Let $w=f(u,v,q)$ with $u=x$, $v=x^2+y^2$, $q=2xy$. At fixed $y$,
$$\boxed{(w_x)_y=f_u+2x f_v+2y f_q.}$$


### Problem 16. Homogeneous-Style Identity
For $z=x^{-1}f(y/x)$, the function is homogeneous of degree $-1$. Euler's theorem gives $xz_x+yz_y=-z$, so
$$\boxed{xz_x+yz_y+z=0.}$$


### Problem 17. Distance to a Surface
Minimize $D^2=x^2+y^2+z^2$ subject to $x-yz=10$. Substituting $x=yz+10$ gives $D^2=(yz+10)^2+y^2+z^2$. The symmetric stationary cases give $y=z=0$ or $y=-z=\pm3$. The latter gives $x=1$ and the smaller value $D^2=19$, so $\boxed{D_{\min}=\sqrt{19}}$ at $(1,3,-3)$ and $(1,-3,3)$.


### Problem 18. Distance to Intersection of Two Planes
Use either vector projection or Lagrange multipliers for $D^2=x^2+y^2+z^2$ with the two plane constraints. Both methods solve the same normal-equation system.


### Problem 19. Maximum Product with Fixed Sum
For positive $x,y,z$ and $x+y+z=1$, Lagrange equations give $1/x=1/y=1/z$, so
$$\boxed{x=y=z=1/3,
\quad xyz\le1/27.}$$


### Problem 20. Cubic on a Cosine Interval
Since $x=\cos\theta$, $-1\le x\le1$. Check critical points of $4x^3+9x^2-12x+3$ and endpoints. The derivative is $12(x+2)(x-1/2)$, so only $x=1/2$ lies inside. Compare values at $-1,1,1/2$.


### Problem 21. Temperature on a Bar
For $0\le x\le5$, $T=4x-x^2$ has critical point $x=2$. Values: $T(0)=0$, $T(2)=4$, $T(5)=-5$. Thus hottest at $x=2$, coldest at $x=5$.


### Problem 22. Temperature on a Region
Check the interior critical point of $T=x^2-y^2-3x$ and the boundary $x=y^2$ plus the limiting boundary as $x\to5^-$. Compare these values to classify hottest and coldest points.


### Problem 23. Differentiating a Sine-Dependent Limit
For $I(t)=\int_0^{\sin t}\sin^{-1}x/x\,dx$,
$$\boxed{I'(t)=\frac{\sin^{-1}(\sin t)}{\sin t}\cos t.}$$


### Problem 24. Hyperbolic Cosine Integral
For $I(x)=\int_{1/x}^{2/x}\cosh(xt)/t\,dt$, substitute $u=xt$ to get $I=\int_1^2\cosh u/u\,du$. Hence $\boxed{I'=0}$.


### Problem 25. Parameter and Limit
Leibniz' rule gives
$$\boxed{\frac{d}{dx}\int_1^{1/x}\frac{e^{xt}}{t}dt
= -\frac{e}{x}+\int_1^{1/x}e^{xt}dt.}$$


### Problem 26. Variable Upper Limit and Parameter
For $I=\int_0^{x^2}\sin(xt)/t\,dt$,
$$\boxed{I'=2x\frac{\sin(x^3)}{x^2}+\int_0^{x^2}\cos(xt)\,dt.}$$


### Problem 27. A Constant Derivative
For $I=\int_{\cos x}^{\sin x}\sqrt{1-t^2}\,dt$,
Leibniz' rule gives
$$I'=\sqrt{1-\sin^2x}\cos x-\sqrt{1-\cos^2x}(-\sin x)=\cos^2x+\sin^2x=\boxed{1}$$
for first-quadrant $x$.


### Problem 28. Maxwell Velocity Form
Using $d(\ln F+\lambda\phi)=0$ gives
$$\frac{f'(x)}{f(x)}+2\lambda x=0,$$
and similarly for $y,z$. Thus $f(x)=A_xe^{-\lambda x^2}$ up to constants, and
$$\boxed{F=Ae^{-(\lambda/2)(x^2+y^2+z^2)}}$$
after renaming the multiplier constant to match the standard convention.


### Problem 29. Heating Estimate
Let $T(t)$ be the given expression. Compute $dT/dt$ by differentiating the upper limit $8/\sqrt t$. At $t=64$, use
$$\Delta t\approx\frac{17-15.73}{T'(64)}.$$
This gives the differential estimate for the waiting time.


### Problem 30. Double Integral Derivative
Let $I(x)=\int_0^x\int_0^x f(s,t)\,ds\,dt$. Then
$$I'=\int_0^x f(x,t)dt+\int_0^x f(s,x)ds.$$
Differentiating again,
$$\boxed{I''=f(x,x)+\int_0^x f_x(x,t)dt+f(x,x)+\int_0^x f_t(s,x)ds.}$$
