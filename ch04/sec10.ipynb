# 4-10. Endpoint or Boundary Point Problems


### Problem 1. Hyperbola Distance
For $x^2-y^2=1$, minimize $D^2=x^2+y^2=1+2y^2$. Hence $y=0$, $x=\pm1$, and $\boxed{D_{\min}=1}$.


### Problem 2. Conic Semiaxes
For $5x^2-6xy+5y^2=32$, diagonalize the quadratic form. Its eigenvalues are $2$ and $8$, so the principal-axis equation is $2u^2+8v^2=32$. Thus the squared distances range from $4$ to $16$, and the semiaxes are $\boxed{4,2}$.


### Problem 3. Second Conic
For $6x^2+4xy+3y^2=28$, the eigenvalues are $\lambda=(9\pm\sqrt{25})/2=7,2$. Thus $2u^2+7v^2=28$, so the semiaxes are
$$\boxed{\sqrt{14}\quad\text{and}\quad 2.}$$


### Problem 4. Quadric Distance
Minimize $x^2+y^2+z^2$ subject to $3x^2+y^2-4xz=4$. Lagrange equations give the nearest point from the smallest generalized eigenvalue of the quadratic constraint.


### Problem 5. Quadric Distance
Use the same Lagrange method on $2z^2+6xy=3$. The minimum occurs when the constraint quadratic form is largest on the unit sphere; then $D^2=3/\lambda_{\max}$.


### Problem 6. Quadric Distance
For $4y^2+2z^2+3xy=18$, diagonalize the symmetric quadratic form and use $D^2=18/\lambda_{\max}$.


### Problem 7. Largest $z$
Use $y=(5-2x)/4$ in $x^2+z^2=2y$. Then
$$z^2=\frac{5-2x}{2}-x^2.$$
Maximizing $z^2$ gives $-1-2x=0$, so $x=-1/2$, $y=3/2$, and
$$\boxed{z_{\max}=\sqrt{7/4}=\frac{\sqrt7}{2}.}$$


### Problem 8. Hottest Point on a Sphere
Maximize $T=xyz$ on $x^2+y^2+z^2=12$. Lagrange equations give $x^2=y^2=z^2=4$. The largest positive temperature occurs at points with either zero or two negative signs:
$$\boxed{T_{\max}=8.}$$


### Problem 9. Temperature on a Disk
Interior critical point: $T_x=4x-2=0$, $T_y=-6y=0$, so $(1/2,0)$ gives $T=-1/2$. On the boundary $x^2+y^2=1$, check with Lagrange multipliers. The hottest point is $(-1,0)$ with $T=4$. On the boundary, $T=5x^2-2x-3$, whose minimum occurs at $x=1/5$; hence the coldest points are $(1/5,\pm 2\sqrt6/5)$ with $T=-16/5$.


### Problem 10. Temperature in a Ball
(a) On $y=0$, $T=xz$ with $x^2+z^2=1$, so $T\in[-1/2,1/2]$.

(b) On the sphere, Lagrange equations for $T=y^2+xz$ give extrema $1$ and $-1/2$.

(c) In the whole ball, the same boundary extrema dominate, with $T=0$ possible inside; hence $\boxed{T_{\max}=1,
T_{\min}=-1/2}$.


### Problem 11. Rectangular Plate
Interior critical point: $(1/2,0)$, value $19/2$. Boundary checks on $x=\pm1$ and $y=\pm1$ give hottest point $(-1,0)$ with $T=14$, and coldest points $(1/2,\pm1)$ with $T=13/2$.


### Problem 12. Sum of Direction Angles with Axes
Let direction cosines be $l,m,n\ge0$ with $l^2+m^2+n^2=1$. The sum $\arccos l+\arccos m+\arccos n$ is symmetric. The interior critical point has $l=m=n=1/\sqrt3$. Boundary checks give the absolute extrema.


### Problem 13. Sum of Angles with Coordinate Planes
The angle with a coordinate plane is the complement of the angle with the corresponding axis. Thus this problem follows from Problem 12 by replacing the sum by $3\pi/2$ minus that sum.


### Problem 14. Example 3b by Lagrange Multipliers
Set $f=x^2+y^2+z^2$ and use the surface constraint from Example 3b. The multiplier equations reproduce the same closest point found by boundary analysis in the text.
