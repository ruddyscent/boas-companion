# 4-9. Maximum and Minimum Problems with Constraints; Lagrange Multipliers


### Problem 1. Rectangle with Isosceles Triangular Ends
Let the rectangle have length $x$, end-triangle height $h$, and total end base $b$. With fixed perimeter, maximize area $A=xb+bh$ subject to the perimeter relation. Lagrange equations give the usual equal-marginal condition: the optimal sloping sides make $60^\circ$ angles with the base.


### Problem 2. Cylinder with a Conical Nose
Let cylinder radius be $r$, cylinder length $h$, and cone height $k$. Maximize $V=\pi r^2h+\pi r^2k/3$ with fixed outside area $S=2\pi rh+\pi r\sqrt{r^2+k^2}+\pi r^2$. The Lagrange equations give the required proportions by solving
$$\nabla V=\lambda\nabla S.$$


### Problem 3. Parcel-Post Box
Let dimensions be $x,y,z$ and take length $x$ with girth $2y+2z$. Constraint: $x+2y+2z=108$. Maximize $V=xyz$. Lagrange equations give
$$\frac1x=\lambda,
\quad \frac1y=2\lambda,
\quad \frac1z=2\lambda,
$$
so $x=2y=2z$. Hence
$$\boxed{x=36,
\quad y=z=18,
\quad V_{\max}=11664\text{ in}^3.}$$


### Problem 4. Box in an Ellipsoid
Maximize $V=8xyz$ with $x^2/4+y^2/9+z^2/25=1$. Lagrange equations give
$$\frac{x^2}{4}=\frac{y^2}{9}=\frac{z^2}{25}=\frac13.
$$
Thus the first-octant vertex is
$$\boxed{(2/\sqrt3,3/\sqrt3,5/\sqrt3)}$$
and the full box has side lengths $4/\sqrt3$, $6/\sqrt3$, $10/\sqrt3$.


### Problem 5. Weighted Distance to a Plane
Minimize $f=4x^2+y^2+z^2$ subject to $2x+3y+z=11$. Lagrange equations give $8x=2\lambda$, $2y=3\lambda$, $2z=\lambda$. Solving with the constraint yields
$$\boxed{(x,y,z)=\left(\frac{11}{31},\frac{66}{31},\frac{22}{31}\right).}$$


### Problem 6. Maximum Coordinate Box Volume
Maximize $V=xyz$ subject to $2x+3y+4z=6$. Lagrange equations give $2x=3y=4z=2$, hence
$$\boxed{x=1,
\quad y=2/3,
\quad z=1/2,
\quad V_{\max}=1/3.}$$


### Problem 7. General Plane Version
For $ax+by+cz=d$, the same calculation gives
$$\boxed{x=d/(3a),\quad y=d/(3b),\quad z=d/(3c),\quad V_{\max}=d^3/(27abc).}$$


### Problem 8. Point on a Line
The sum of squared distances from $(x,y)$ to $(1,0)$ and $(-1,0)$ is
$$S=(x-1)^2+y^2+(x+1)^2+y^2=2x^2+2y^2+2.
$$
Minimize this subject to $2x+3y=4$. The closest point on the line to the origin is
$$\boxed{(x,y)=\left(\frac8{13},\frac{12}{13}\right).}$$


### Problem 9. Triangle in an Ellipse
For a symmetric triangle with top vertex $(a\cos\theta,0)$ and base endpoints $(-a\cos\theta,\pm b\sin\theta)$, area is $A=2ab\sin\theta\cos\theta$. This is maximized at $\theta=\pi/4$. Thus the largest area is
$$\boxed{A_{\max}=ab.}$$


### Problem 10. Completing the Example
Apply $\nabla f=\lambda\nabla g$ to the example's objective and constraint. The multiplier equations reproduce the same stationary point obtained by elimination, with the same maximum/minimum classification from the geometry.


### Problem 11. Distance to Intersection of Two Planes
The line satisfies $2x+y-z=1$ and $x-y+z=2$. Minimize $x^2+y^2+z^2$. Lagrange equations give
$$2(x,y,z)=\lambda(2,1,-1)+\mu(1,-1,1).$$
Solving with the two constraints gives the line $(x,y,z)=(1,t,t+1)$. Minimizing $1+t^2+(t+1)^2$ gives $t=-1/2$, so
$$\boxed{(x,y,z)=\left(1,-\frac12,\frac12\right),\qquad d_{\min}=\sqrt{\frac32}.}$$


### Problem 12. Right Triangular Prism
Let the right-triangle legs be $a,b$ and prism length $l$. With fixed volume $V=abl/2$, minimize area
$$S=ab+l(a+b+\sqrt{a^2+b^2}).$$
Symmetry gives $a=b$. Substitution then determines $l$ from $V$; the least-area prism has an isosceles right-triangle base.
