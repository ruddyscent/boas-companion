# 4-12. Differentiation of Integrals; Leibniz' Rule


### Problem 1. Variable Upper Limit
$$\boxed{dy/dx=\sin(x)/(2\sqrt{x}).}$$


### Problem 2. Partials with Respect to Limits
If $s=\int_u^v(1-e^t)/t\,dt$, then
$$\boxed{s_v=(1-e^v)/v,
\quad s_u=-(1-e^u)/u.}$$
Both limits at zero are $-1$ and $+1$, respectively.


### Problem 3. Both Limits Variable
$$\boxed{z'=\frac{\sin(\cos x)}{\cos x}(-\sin x)-\frac{\sin(\sin x)}{\sin x}\cos x.}$$


### Problem 4. Limit by L'Hopital
The expression is the average value of $\sin t/t$ on $[2,x]$, so the limit is
$$\boxed{\sin2/2.}$$


### Problem 5. Integral with Two Variables
For $u=\int_x^{y-x}\sin t/t\,dt$,
$$\boxed{u_x=-\frac{\sin(y-x)}{y-x}-\frac{\sin x}{x},
\quad u_y=\frac{\sin(y-x)}{y-x}.}$$
At $x=\pi/2$, $y=\pi$, this gives $u_x=-4/\pi$, $u_y=2/\pi$.


### Problem 6. Logarithmic Integral Limits
For $w=\int_{xy}^{2x-3y}du/\ln u$,
$$\boxed{w_x=\frac{2}{\ln(2x-3y)}-\frac{y}{\ln(xy)},
\quad w_y=\frac{-3}{\ln(2x-3y)}-\frac{x}{\ln(xy)}.}$$
At $(3,1)$ both limits are $3$, so the formulas are evaluated by their limiting common endpoint behavior.


### Problem 7. Implicit Integral System
Differentiate $\int_u^v e^{-t^2}dt=x$ and $uv=y$. At $u=2$, $v=0$, solve
$$-e^{-4}du+dv=dx,
\qquad vdu+udv=dy.$$
This gives the requested partials directly.


### Problem 8. Inverse Function from an Integral
If $u=\int_0^x e^{-s^2}ds$, then
$$\boxed{dx/du=e^{x^2}.}$$


### Problem 9. Parameter in the Integrand
Evaluate first:
$$y=\int_0^\pi\sin(xt)dt=\frac{1-\cos(\pi x)}{x}.$$
Differentiate this, or differentiate under the integral sign:
$$\boxed{y'=\int_0^\pi t\cos(xt)dt.}$$
The two forms agree.


### Problem 10. Exponential Integral Identity
Since $(e^{xu}-1)/u=\int_0^x e^{su}ds$, differentiating under the integral gives
$$\boxed{y'=\int_0^1 e^{xu}du=\frac{e^x-1}{x}.}$$


### Problem 11. Evaluate First or Differentiate First
For $I=\int_{3-x}^{x^2}(x-t)dt$, Leibniz' rule gives
$$\boxed{I'=\int_{3-x}^{x^2}1\,dt+(x-x^2)2x-(x-(3-x))(-1).}$$


### Problem 12. Parameter and Variable Limits
For $I=\int_x^{x^2}du/\ln(x+u)$,
$$\boxed{I'=\frac{2x}{\ln(x+x^2)}-\frac1{\ln(2x)}+\int_x^{x^2}\frac{-1}{(x+u)[\ln(x+u)]^2}\,du.}$$


### Problem 13. Scaled Limits
For $I=\int_{1/x}^{2/x}\sin(xt)/t\,dt$, substitute $u=xt$ to see
$$I=\int_1^2\frac{\sin u}{u}\,du,$$
so $\boxed{I'=0}$.


### Problem 14. Differentiating a Known Integral
From $\int_0^\infty dx/(y^2+x^2)=\pi/(2y)$,
$$-2y\int_0^\infty\frac{dx}{(y^2+x^2)^2}=-\frac{\pi}{2y^2}.
$$
Thus
$$\boxed{\int_0^\infty\frac{dx}{(y^2+x^2)^2}=\frac{\pi}{4y^3}.}$$


### Problem 15. Two Parameter Derivatives
Differentiate $k/(a^2+k^2)$ with respect to $a$ and $k$:
$$\boxed{\int_0^\infty xe^{-ax}\sin kx\,dx=\frac{2ka}{(a^2+k^2)^2},}$$
$$\boxed{\int_0^\infty xe^{-ax}\cos kx\,dx=\frac{a^2-k^2}{(a^2+k^2)^2}.}$$


### Problem 16. Gaussian Moments
From $I_0(a)=\frac12\sqrt{\pi/a}$ and $I_{2m}=(-1)^m d^mI_0/da^m$,
$$\boxed{\int_0^\infty t^{2m}e^{-at^2}dt=\frac{(2m-1)!!}{2^{m+1}}\sqrt\pi\,a^{-m-1/2}.}$$

### Problem 18. Poisson Kernel Harmonicity
Differentiate under the integral sign twice. The kernel $y/[(x-t)^2+y^2]$ satisfies $K_{xx}+K_{yy}=0$, so
$$\boxed{u_{xx}+u_{yy}=0.}$$


### Problem 19. Sine-Convolution ODE
Differentiate
$$y=\int_0^x f(u)\sin(x-u)du.$$
Then $y'=\int_0^x f(u)\cos(x-u)du$ and $y''=f(x)-\int_0^x f(u)\sin(x-u)du$, so
$$\boxed{y''+y=f(x).}$$


### Problem 20. Repeated Integral Kernel
(a) Substitute $u=x-t$ to get $y=\int_0^x f(u)du$, so $y'=f(x)$.

(b) Differentiating once gives $y'=\int_0^x f(u)du$, so $y''=f(x)$.

(c) Repeated differentiation lowers the power of $(x-u)$ one step at a time, giving
$$\boxed{y^{(n)}=f(x).}$$
