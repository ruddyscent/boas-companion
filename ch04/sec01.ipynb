## Basic Partial Derivatives: Problems 1 through 6


### Problem 1. Quotient in Two Variables

Let

$$
u=\frac{x^2}{x^2+y^2}.
$$

Using the quotient rule,

$$
\frac{\partial u}{\partial x}
=\frac{2x(x^2+y^2)-x^2(2x)}{(x^2+y^2)^2}
=\frac{2xy^2}{(x^2+y^2)^2}.
$$

Similarly,

$$
\frac{\partial u}{\partial y}
=\frac{0\cdot (x^2+y^2)-x^2(2y)}{(x^2+y^2)^2}
=-\frac{2x^2y}{(x^2+y^2)^2}.
$$

Thus, away from $(0,0)$,

$$
\boxed{u_x=\frac{2xy^2}{(x^2+y^2)^2},\qquad
u_y=-\frac{2x^2y}{(x^2+y^2)^2}.}
$$


### Problem 2. Power Function with Variable Exponent

For

$$
s=t^u,
$$

hold the other variable constant in each differentiation. With respect to $t$, use the usual power rule:

$$
\frac{\partial s}{\partial t}=ut^{u-1}.
$$

With respect to $u$, write $t^u=e^{u\ln t}$, so

$$
\frac{\partial s}{\partial u}=t^u\ln t.
$$

Thus, for $t>0$,

$$
\boxed{\frac{\partial s}{\partial t}=ut^{u-1},\qquad
\frac{\partial s}{\partial u}=t^u\ln t.}
$$


### Problem 3. Logarithm of a Radial Quantity

Write

$$
z=\ln\sqrt{u^2+v^2+w^2}
=\frac12\ln(u^2+v^2+w^2).
$$

Therefore

$$
\frac{\partial z}{\partial u}
=\frac12\frac{2u}{u^2+v^2+w^2}
=\frac{u}{u^2+v^2+w^2}.
$$

The other variables are symmetric, so

$$
\boxed{
z_u=\frac{u}{u^2+v^2+w^2},\qquad
z_v=\frac{v}{u^2+v^2+w^2},\qquad
z_w=\frac{w}{u^2+v^2+w^2}.}
$$


### Problem 4. Second Partials at Critical Points

For

$$
w=x^3-y^3-2xy+6,
$$

the first partial derivatives are

$$
w_x=3x^2-2y,
\qquad
w_y=-3y^2-2x.
$$

Set both equal to zero. From $3x^2=2y$ and $-3y^2=2x$,

$$
y=\frac32x^2,
\qquad
x=-\frac32y^2.
$$

Substitution gives

$$
x=-\frac32\left(\frac32x^2\right)^2=-\frac{27}{8}x^4,
$$

so

$$
x\left(1+\frac{27}{8}x^3\right)=0.
$$

Thus the critical points are

$$
(0,0),
\qquad
\left(-\frac23,\frac23\right).
$$

The requested second partials are

$$
w_{xx}=6x,
\qquad
w_{yy}=-6y.
$$

At the two points,

$$
\boxed{(0,0):\quad w_{xx}=0,
\quad w_{yy}=0,}
$$

and

$$
\boxed{\left(-\frac23,\frac23\right):\quad w_{xx}=-4,
\quad w_{yy}=-4.}
$$


### Problem 5. Quartic Critical Points

For

$$
w=8x^4+y^4-2xy^2,
$$

the first partial derivatives are

$$
w_x=32x^3-2y^2,
\qquad
w_y=4y^3-4xy=4y(y^2-x).
$$

The equations $w_x=w_y=0$ give

$$
y^2=16x^3,
\qquad
y=0\quad\text{or}\quad x=y^2.
$$

If $y=0$, then $x=0$. If $x=y^2$, then

$$
y^2=16y^6,
$$

so either $y=0$ again, or $y^4=1/16$. Hence the critical points are

$$
(0,0),
\qquad
\left(\frac14,\frac12\right),
\qquad
\left(\frac14,-\frac12\right).
$$

The requested second partials are

$$
w_{xx}=96x^2,
\qquad
w_{yy}=12y^2-4x.
$$

Therefore

$$
\boxed{(0,0):\quad w_{xx}=0,
\quad w_{yy}=0,}
$$

and at both nonzero critical points,

$$
\boxed{\left(\frac14,\pm\frac12\right):\quad w_{xx}=6,
\quad w_{yy}=2.}
$$


### Problem 6. Equality of Mixed Partials and Laplace Equation

Let

$$
u=e^x\cos y.
$$

The first partial derivatives are

$$
u_x=e^x\cos y,
\qquad
u_y=-e^x\sin y.
$$

For part (a),

$$
u_{xy}=\frac{\partial}{\partial y}(e^x\cos y)=-e^x\sin y,
$$

and

$$
u_{yx}=\frac{\partial}{\partial x}(-e^x\sin y)=-e^x\sin y.
$$

Thus

$$
\boxed{u_{xy}=u_{yx}.}
$$

For part (b),

$$
u_{xx}=e^x\cos y,
\qquad
u_{yy}=-e^x\cos y,
$$

so

$$
\boxed{u_{xx}+u_{yy}=0.}
$$


## Coordinate-Dependent Partials: Problems 7 through 24

For this group,

$$
z=x^2+2y^2,
\qquad
x=r\cos\theta,
\qquad
y=r\sin\theta.
$$

The value of a partial derivative depends on which second variable is held fixed, so each problem rewrites $z$ only in the variables needed for that derivative.


### Problem 7. $x$-Partial with $y$ Fixed

Since $y$ is fixed, the original expression is already written in the right variables:

$$
z=x^2+2y^2.
$$

$$
\boxed{\left(\frac{\partial z}{\partial x}\right)_y=2x.}
$$

### Problem 8. $x$-Partial with $r$ Fixed

Here $r$ is fixed, so use

$$
r^2=x^2+y^2
\quad\Rightarrow\quad
y^2=r^2-x^2.
$$

Then

$$
z=x^2+2(r^2-x^2)=2r^2-x^2.
$$

$$
\boxed{\left(\frac{\partial z}{\partial x}\right)_r=-2x.}
$$

### Problem 9. $x$-Partial with $\theta$ Fixed

Here $\theta$ is fixed. Since

$$
\frac{y}{x}=\tan\theta,
\qquad
y=x\tan\theta,
$$

we get

$$
z=x^2+2x^2\tan^2\theta=x^2(1+2\tan^2\theta).
$$

$$
\boxed{\left(\frac{\partial z}{\partial x}\right)_\theta
=2x(1+2\tan^2\theta).}
$$

### Problem 10. $y$-Partial with $x$ Fixed

Since $x$ is fixed, keep $z$ as a function of $x$ and $y$:

$$
z=x^2+2y^2.
$$

$$
\boxed{\left(\frac{\partial z}{\partial y}\right)_x=4y.}
$$

### Problem 11. $y$-Partial with $r$ Fixed

Here $r$ is fixed, so

$$
x^2=r^2-y^2.
$$

Thus

$$
z=(r^2-y^2)+2y^2=r^2+y^2.
$$

$$
\boxed{\left(\frac{\partial z}{\partial y}\right)_r=2y.}
$$

### Problem 12. $y$-Partial with $\theta$ Fixed

Here $\theta$ is fixed. Since

$$
\frac{x}{y}=\cot\theta,
\qquad
x=y\cot\theta,
$$

we get

$$
z=y^2\cot^2\theta+2y^2=y^2(\cot^2\theta+2).
$$

$$
\boxed{\left(\frac{\partial z}{\partial y}\right)_\theta
=2y(\cot^2\theta+2).}
$$


### Problem 13. $\theta$-Partial with $x$ Fixed

With $x$ fixed, write $y=x\tan\theta$. Then

$$
z=x^2+2x^2\tan^2\theta=x^2(1+2\tan^2\theta).
$$

$$
\boxed{\left(\frac{\partial z}{\partial\theta}\right)_x
=4x^2\tan\theta\sec^2\theta.}
$$

### Problem 14. $\theta$-Partial with $y$ Fixed

With $y$ fixed, write $x=y\cot\theta$. Then

$$
z=y^2\cot^2\theta+2y^2=y^2(\cot^2\theta+2).
$$

$$
\boxed{\left(\frac{\partial z}{\partial\theta}\right)_y
=-2y^2\cot\theta\csc^2\theta.}
$$

### Problem 15. $\theta$-Partial with $r$ Fixed

With $r$ fixed, substitute $x=r\cos\theta$ and $y=r\sin\theta$:

$$
z=r^2\cos^2\theta+2r^2\sin^2\theta
=r^2(\cos^2\theta+2\sin^2\theta).
$$

$$
\boxed{\left(\frac{\partial z}{\partial\theta}\right)_r
=2r^2\sin\theta\cos\theta.}
$$

### Problem 16. $r$-Partial with $\theta$ Fixed

With $\theta$ fixed, again substitute $x=r\cos\theta$ and $y=r\sin\theta$:

$$
z=r^2\cos^2\theta+2r^2\sin^2\theta
=r^2(\cos^2\theta+2\sin^2\theta).
$$

$$
\boxed{\left(\frac{\partial z}{\partial r}\right)_\theta
=2r(\cos^2\theta+2\sin^2\theta).}
$$

### Problem 17. $r$-Partial with $x$ Fixed

With $x$ fixed, use $y^2=r^2-x^2$. Then

$$
z=x^2+2(r^2-x^2)=2r^2-x^2.
$$

$$
\boxed{\left(\frac{\partial z}{\partial r}\right)_x=4r.}
$$

### Problem 18. $r$-Partial with $y$ Fixed

With $y$ fixed, use $x^2=r^2-y^2$. Then

$$
z=(r^2-y^2)+2y^2=r^2+y^2.
$$

$$
\boxed{\left(\frac{\partial z}{\partial r}\right)_y=2r.}
$$


### Problem 19. Mixed Partial in $(r,y)$

Use the variables $(y,r)$ because $\partial^2z/\partial r\,\partial y$ means differentiate first with respect to $y$, then with respect to $r$. Since

$$
x^2=r^2-y^2,
$$

we have

$$
z=(r^2-y^2)+2y^2=r^2+y^2.
$$

Thus $z_y=2y$, and differentiating with respect to $r$ gives

$$
\boxed{\frac{\partial^2z}{\partial r\,\partial y}=0.}
$$

### Problem 20. Mixed Partial in $(x,\theta)$

Use the variables $(x,\theta)$. With $\theta$ fixed relative to $x$, write $y=x\tan\theta$, so

$$
z=x^2+2x^2\tan^2\theta=x^2(1+2\tan^2\theta).
$$

$$
\frac{\partial z}{\partial\theta}=4x^2\tan\theta\sec^2\theta,
$$

so

$$
\boxed{\frac{\partial^2z}{\partial x\,\partial\theta}
=8x\tan\theta\sec^2\theta.}
$$

### Problem 21. Mixed Partial in $(y,\theta)$

Use the variables $(y,\theta)$. With $\theta$ fixed relative to $y$, write $x=y\cot\theta$, so

$$
z=y^2\cot^2\theta+2y^2=y^2(\cot^2\theta+2).
$$

$$
\frac{\partial z}{\partial\theta}=-2y^2\cot\theta\csc^2\theta,
$$

so

$$
\boxed{\frac{\partial^2z}{\partial y\,\partial\theta}
=-4y\cot\theta\csc^2\theta.}
$$

### Problem 22. Mixed Partial in $(r,x)$

Use the variables $(x,r)$. Since

$$
y^2=r^2-x^2,
$$

we get

$$
z=x^2+2(r^2-x^2)=2r^2-x^2.
$$

Then $z_x=-2x$, and differentiating with respect to $r$ gives

$$
\boxed{\frac{\partial^2z}{\partial r\,\partial x}=0.}
$$

### Problem 23. Mixed Partial in $(r,\theta)$

Use the variables $(r,\theta)$. Substitute $x=r\cos\theta$ and $y=r\sin\theta$:

$$
z=r^2\cos^2\theta+2r^2\sin^2\theta
=r^2(\cos^2\theta+2\sin^2\theta).
$$

$$
\frac{\partial z}{\partial\theta}=2r^2\sin\theta\cos\theta,
$$

so

$$
\boxed{\frac{\partial^2z}{\partial r\,\partial\theta}
=4r\sin\theta\cos\theta.}
$$

### Problem 24. Mixed Partial in $(x,y)$

Use the original variables $(x,y)$:

$$
z=x^2+2y^2.
$$

Then $z_y=4y$, and differentiating with respect to $x$ gives

$$
\boxed{\frac{\partial^2z}{\partial x\,\partial y}=0.}
$$


## Repeated Coordinate Partials: Problems 7' through 24'

For this group,

$$
z=r^2\tan^2\theta.
$$

As in Problems 7 through 24, each problem rewrites $z$ only in the variables needed for that derivative. The formulas below are used where the displayed denominators are nonzero.


### Problem 7'. $x$-Partial with $y$ Fixed

Since $y$ is fixed, write $z$ in terms of $x$ and $y$. From

$$
r^2=x^2+y^2,
\qquad
\tan^2\theta=\frac{y^2}{x^2},
$$

we get

$$
z=(x^2+y^2)\frac{y^2}{x^2}
=y^2+\frac{y^4}{x^2}.
$$

$$
\boxed{\left(\frac{\partial z}{\partial x}\right)_y
=-\frac{2y^4}{x^3}.}
$$

### Problem 8'. $x$-Partial with $r$ Fixed

Here $r$ is fixed. Since $y^2=r^2-x^2$,

$$
\tan^2\theta=\frac{y^2}{x^2}
=\frac{r^2-x^2}{x^2}.
$$

Thus

$$
z=r^2\frac{r^2-x^2}{x^2}
=\frac{r^4}{x^2}-r^2.
$$

$$
\boxed{\left(\frac{\partial z}{\partial x}\right)_r
=-\frac{2r^4}{x^3}.}
$$

### Problem 9'. $x$-Partial with $\theta$ Fixed

Here $\theta$ is fixed. Since $x=r\cos\theta$,

$$
r=x\sec\theta.
$$

Hence

$$
z=x^2\sec^2\theta\tan^2\theta.
$$

$$
\boxed{\left(\frac{\partial z}{\partial x}\right)_\theta
=2x\sec^2\theta\tan^2\theta.}
$$

### Problem 10'. $y$-Partial with $x$ Fixed

Since $x$ is fixed, write $z$ in terms of $x$ and $y$:

$$
z=(x^2+y^2)\frac{y^2}{x^2}
=y^2+\frac{y^4}{x^2}.
$$

$$
\boxed{\left(\frac{\partial z}{\partial y}\right)_x
=2y+\frac{4y^3}{x^2}.}
$$

### Problem 11'. $y$-Partial with $r$ Fixed

Here $r$ is fixed. Since $x^2=r^2-y^2$,

$$
\tan^2\theta=\frac{y^2}{x^2}
=\frac{y^2}{r^2-y^2}.
$$

Therefore

$$
z=\frac{r^2y^2}{r^2-y^2}.
$$

$$
\left(\frac{\partial z}{\partial y}\right)_r
=r^2\frac{2y(r^2-y^2)+2y^3}{(r^2-y^2)^2}
=\frac{2r^4y}{(r^2-y^2)^2}.
$$

Thus

$$
\boxed{\left(\frac{\partial z}{\partial y}\right)_r
=\frac{2r^4y}{(r^2-y^2)^2}.}
$$

### Problem 12'. $y$-Partial with $\theta$ Fixed

Here $\theta$ is fixed. Since $y=r\sin\theta$,

$$
r=y\csc\theta.
$$

Thus

$$
z=y^2\csc^2\theta\tan^2\theta
=y^2\sec^2\theta.
$$

$$
\boxed{\left(\frac{\partial z}{\partial y}\right)_\theta
=2y\sec^2\theta.}
$$


### Problem 13'. $\theta$-Partial with $x$ Fixed

With $x$ fixed, use $r=x\sec\theta$. Then

$$
z=x^2\sec^2\theta\tan^2\theta.
$$

$$
\frac{d}{d\theta}(\sec^2\theta\tan^2\theta)
=2\tan\theta\sec^2\theta(1+2\tan^2\theta),
$$

so

$$
\boxed{\left(\frac{\partial z}{\partial\theta}\right)_x
=2x^2\tan\theta\sec^2\theta(1+2\tan^2\theta).}
$$

### Problem 14'. $\theta$-Partial with $y$ Fixed

With $y$ fixed, use $r=y\csc\theta$. Then

$$
z=y^2\csc^2\theta\tan^2\theta
=y^2\sec^2\theta.
$$

$$
\boxed{\left(\frac{\partial z}{\partial\theta}\right)_y
=2y^2\sec^2\theta\tan\theta.}
$$

### Problem 15'. $\theta$-Partial with $r$ Fixed

With $r$ fixed, the expression is already in the variables $(r,\theta)$:

$$
z=r^2\tan^2\theta.
$$

$$
\boxed{\left(\frac{\partial z}{\partial\theta}\right)_r
=2r^2\tan\theta\sec^2\theta.}
$$

### Problem 16'. $r$-Partial with $\theta$ Fixed

With $\theta$ fixed, the expression is again already in the variables $(r,\theta)$:

$$
z=r^2\tan^2\theta.
$$

$$
\boxed{\left(\frac{\partial z}{\partial r}\right)_\theta
=2r\tan^2\theta.}
$$

### Problem 17'. $r$-Partial with $x$ Fixed

With $x$ fixed, use $y^2=r^2-x^2$. Then

$$
z=r^2\frac{r^2-x^2}{x^2}
=\frac{r^4}{x^2}-r^2.
$$

$$
\boxed{\left(\frac{\partial z}{\partial r}\right)_x
=\frac{4r^3}{x^2}-2r.}
$$

### Problem 18'. $r$-Partial with $y$ Fixed

With $y$ fixed, use $x^2=r^2-y^2$. Then

$$
z=r^2\frac{y^2}{r^2-y^2}
=\frac{r^2y^2}{r^2-y^2}.
$$

$$
\left(\frac{\partial z}{\partial r}\right)_y
=y^2\frac{2r(r^2-y^2)-2r^3}{(r^2-y^2)^2}
=-\frac{2ry^4}{(r^2-y^2)^2}.
$$

Thus

$$
\boxed{\left(\frac{\partial z}{\partial r}\right)_y
=-\frac{2ry^4}{(r^2-y^2)^2}.}
$$


### Problem 19'. Mixed Partial in $(r,y)$

Use the variables $(y,r)$. Since $x^2=r^2-y^2$,

$$
z=r^2\frac{y^2}{r^2-y^2}
=\frac{r^2y^2}{r^2-y^2}.
$$

$$
\frac{\partial z}{\partial y}=\frac{2r^4y}{(r^2-y^2)^2}.
$$

Differentiating this with respect to $r$ gives

$$
\boxed{\frac{\partial^2z}{\partial r\,\partial y}
=-\frac{8r^3y^3}{(r^2-y^2)^3}.}
$$

### Problem 20'. Mixed Partial in $(x,\theta)$

Use the variables $(x,\theta)$. With $x$ fixed, $r=x\sec\theta$, so

$$
z=x^2\sec^2\theta\tan^2\theta.
$$

$$
\frac{\partial z}{\partial\theta}
=2x^2\tan\theta\sec^2\theta(1+2\tan^2\theta).
$$

Differentiating this with respect to $x$ gives

$$
\boxed{\frac{\partial^2z}{\partial x\,\partial\theta}
=4x\tan\theta\sec^2\theta(1+2\tan^2\theta).}
$$

### Problem 21'. Mixed Partial in $(y,\theta)$

Use the variables $(y,\theta)$. With $y$ fixed, $r=y\csc\theta$, so

\frac{\partial z}{\partial\theta}=2y^2\sec^2\theta\tan\theta.

### Problem 22'. Mixed Partial in $(r,x)$
