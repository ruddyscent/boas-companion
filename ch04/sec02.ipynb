# 4-2. Power Series in Two Variables


### Problem 1. Product Series
Use $\cos x=\sum_{m\ge0}(-1)^m x^{2m}/(2m)!$ and $\sinh y=\sum_{n\ge0}y^{2n+1}/(2n+1)!$. Then
$$
\boxed{\cos x\sinh y=\sum_{m,n\ge0}\frac{(-1)^m x^{2m}y^{2n+1}}{(2m)!(2n+1)!}.}
$$
The first terms are $y-x^2y/2+y^3/6+x^4y/24-x^2y^3/12+\cdots$.


### Problem 2. Cosine of a Sum
Substitute $x+y$ into the one-variable series:
$$
\boxed{\cos(x+y)=\sum_{n=0}^\infty(-1)^n\frac{(x+y)^{2n}}{(2n)!}.}
$$
Thus $\cos(x+y)=1-(x^2+2xy+y^2)/2+(x+y)^4/24-\cdots$.


### Problem 3. Logarithm of a Quotient
Write $\ln[(1+x)/(1+y)]=\ln(1+x)-\ln(1+y)$. Therefore
$$
\boxed{\ln\frac{1+x}{1+y}=\sum_{n=1}^\infty(-1)^{n+1}\frac{x^n-y^n}{n}.}
$$


### Problem 4. Exponential of a Product
Use $u=xy$ in $e^u$:
$$
\boxed{e^{xy}=\sum_{n=0}^\infty\frac{x^ny^n}{n!}.}
$$


### Problem 5. Binomial Square Root
The binomial series gives
$$
\boxed{\sqrt{1+xy}=1+\frac12xy-\frac18x^2y^2+\frac1{16}x^3y^3-\frac5{128}x^4y^4+\cdots.}
$$


### Problem 6. Exponential of a Sum
Since $e^{x+y}=e^xe^y$,
$$
\boxed{e^{x+y}=\sum_{m,n\ge0}\frac{x^my^n}{m!n!}.}
$$


### Problem 7. Third-Order Taylor Coefficients
Let $h=x-a$ and $k=y-b$. The third-order part is
$$
\frac1{3!}\left(h\partial_x+k\partial_y\right)^3f(a,b)
=\boxed{\frac{f_{xxx}}6h^3+\frac{f_{xxy}}2h^2k+\frac{f_{xyy}}2hk^2+\frac{f_{yyy}}6k^3.}
$$
This matches the $n=3$ terms in the two-variable Taylor formula.


### Problem 8. Real and Imaginary Parts
From $e^{x+iy}=e^x(\cos y+i\sin y)$,
$$
e^x\cos y=\Re\sum_{n=0}^\infty\frac{(x+iy)^n}{n!},\qquad
 e^x\sin y=\Im\sum_{n=0}^\infty\frac{(x+iy)^n}{n!}.
$$
Thus
$$
\boxed{e^x\cos y=1+x+\frac{x^2-y^2}{2}+\frac{x^3-3xy^2}{6}+\frac{x^4-6x^2y^2+y^4}{24}+\cdots,}
$$
$$
\boxed{e^x\sin y=y+xy+\frac{x^2y}{2}-\frac{y^3}{6}+\frac{x^3y}{6}-\frac{xy^3}{6}+\cdots.}
$$
