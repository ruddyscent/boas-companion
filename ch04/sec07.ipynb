# 4-7. More Chain Rule


### Problem 1. Parametric Implicit Derivatives
From $x=yz$ and $y=2\sin(y+z)$, treat $z=z(y)$. Then $1=2\cos(y+z)(1+z')$, so $z'=\tfrac12\sec(y+z)-1$. Hence
$$\boxed{dx/dy=z+yz'.}$$
Differentiate again: $z''=\tfrac12\sec(y+z)\tan(y+z)(1+z')$, and $\boxed{d^2x/dy^2=2z'+yz''}$.


### Problem 2. One-Variable Constraint
For $P=r\cos t$ and $r\sin t-2te^r=0$, implicit differentiation gives
$$
r'\sin t+r\cos t-2e^r-2te^r r'=0.
$$
Thus $r'=(2e^r-r\cos t)/(\sin t-2te^r)$, and
$$\boxed{P'=r'\cos t-r\sin t.}$$


### Problem 3. Two Independent Variables
For $z=xe^{-y}$, $x=\cosh t$, $y=\cos s$,
$$\boxed{z_s=xe^{-y}\sin s,\qquad z_t=e^{-y}\sinh t.}$$


### Problem 4. Composite Exponential
For $w=e^{-r^2-s^2}$, $r=uv$, $s=u+2v$,
$$\boxed{w_u=w[-2r v-2s],\qquad w_v=w[-2r u-4s].}$$


### Problem 5. Three-Factor Composite
Use logarithmic differentiation:
$$
\ln u=2\ln x+3\ln y+\ln z.
$$
With $x=\sin(s+t)$, $y=\cos(s+t)$, $z=e^{st}$,
$$
\boxed{u_s=u\{2\cot(s+t)-3\tan(s+t)+t\},}
$$
$$
\boxed{u_t=u\{2\cot(s+t)-3\tan(s+t)+s\}.}
$$


### Problem 6. Polar Derivative Operators
For $w=f(x,y)$, $x=r\cos\theta$, $y=r\sin\theta$,
$$\boxed{w_r=w_x\cos\theta+w_y\sin\theta,}
$$
$$\boxed{w_\theta=-rw_x\sin\theta+rw_y\cos\theta.}
$$
Also
$$\boxed{w_{rr}=w_{xx}\cos^2\theta+2w_{xy}\sin\theta\cos\theta+w_{yy}\sin^2\theta.}
$$


### Problem 7. Holding Different Variables Fixed
From $y=r\sin\theta$,
$$\boxed{(\partial y/\partial\theta)_r=r\cos\theta.}
$$
If $x$ is fixed, $0=\cos\theta\,dr-r\sin\theta\,d\theta$, so $dr=r\tan\theta\,d\theta$ and
$$\boxed{(\partial y/\partial\theta)_x=r\sec\theta.}
$$
Also $\theta=\tan^{-1}(y/x)$ gives $\boxed{(\partial\theta/\partial y)_x=x/(x^2+y^2)}$. Reciprocals occur only when the same variable is held fixed.


### Problem 8. Linearized System at a Point
Differentiate the two given equations and substitute $(x,y,s,t)=(1,-3,2,-1)$. The resulting two linear equations determine each requested partial. Solving them gives
$$\boxed{x_s=-\frac{19}{13},\quad y_s=\frac{24}{13},\quad x_t=-\frac{21}{13},\quad y_t=\frac{6}{13}.}$$


### Problem 9. Verification of Polar Inverse Formulas
All three methods give
$$\boxed{r_x=\cos\theta,
\quad r_y=\sin\theta,
\quad \theta_x=-\frac{\sin\theta}{r},
\quad \theta_y=\frac{\cos\theta}{r}.}
$$


### Problem 10. Another Linearized System
Differentiating the two equations and substituting $(x,y,s,t)=(4,2,5,3)$ gives a $2\times2$ linear system for each held variable. Solving gives
$$\boxed{x_s=\frac16,\quad y_s=\frac76,\quad x_t=\frac{13}{6},\quad y_t=-\frac{11}{6}.}$$


### Problem 11. Completing the Example
The general answer is obtained by solving the linear differential system for $dx$ and $dy$, then substituting into $dz=dx-dy$. The numerical values follow by substituting before solving; this gives the same result with less algebra.


### Problem 12. Derivatives of $w=x+y$
From $dw=dx+dy$ and the two constraint differentials,
$$
(3x^2+y)dx+(x+3y^2)dy=ds,
$$
$$
(2xy+y^2)dx+(x^2+2xy)dy=dt.
$$
Solve this linear system for $dx+dy$ to obtain $w_s$ and $w_t$.


### Problem 13. Thermodynamic-Style Partials
Use $dm=qdp+pdq$ and differentiate the two defining equations. For example, at fixed $m$, $qdp+pdq=0$, so $\boxed{(\partial p/\partial q)_m=-p/q}$. The remaining requested derivatives follow by setting the named variable constant in the same differential system.


### Problem 14. Partial with $z$ Fixed
Use $du=(2x+yz)dx+(2y+xz)dy+xy\,dz$ and differentiate the constraint with $dz=0$. At $(2,1,1)$ this gives $dy/dx=-14/3$. Hence
$$\boxed{(\partial u/\partial x)_z=5+5(-14/3)=-55/3.}$$


### Problem 15. Two Ways to Hold Variables
Differentiate
$$x^2u-y^2v=1,\qquad x+y=uv.$$
Set $dv=0$ for $(\partial x/\partial u)_v$, and set $dy=0$ for $(\partial x/\partial u)_y$; solve the resulting two linear equations in each case.


### Problem 16. Three Subproblems for $w$
Use $dw=(2x+y)dx+xdy+2zdz$. In each part, differentiate the listed constraints, solve for the needed differentials, and substitute into $dw$.


### Problem 17. Partial Derivatives at a Point
Differentiate the two equations for $p,q,s,t$. At $(-1,2,3,5)$, solve the two linear systems obtained by holding $t$ and then $q$ fixed. This gives the requested values.


### Problem 18. Variables $m,n$ from $a,b$
From $dm=da+db$ and $dn=2ada+2bdb$. Holding $n$ fixed gives $ada+bdb=0$, so
$$\boxed{(\partial b/\partial m)_n=\frac{a}{a-b}.}$$
Also, holding $a$ fixed gives $dm=db$, hence $\boxed{(\partial m/\partial b)_a=1}$.


### Problem 19. Derivatives with $z,r,s,y$ Held Fixed
Use $dz=dr+2sds$ and the two constraint differentials. Substitute $(-1,2,3,1,3)$, impose the indicated held variable, and solve the resulting linear equations.


### Problem 20. Four Partials from Two Constraints
Differentiate the two equations in $u,v,x,y$ and substitute $(2,-1,3,2)$. For each requested partial, set the named held differential to zero and solve the corresponding $2\times2$ system.


### Problem 21. Six Partials at a Point
For the sphere, $2xdx+2ydy+2zdz=0$. For the second constraint, $3w^2dw+3z^2dz=5y dx+5x dy$. Substitute $(1,-2,1,1)$ and impose each held variable to solve for the requested derivative.


### Problem 22. Composite One-Variable Form
Let $q=ax+by$ and $w=f(q)$. Then $w_x=af'(q)$ and $w_y=bf'(q)$, so
$$\boxed{b w_x-a w_y=0.}$$


### Problem 23. Wave Equation Form
For $u=f(x-ct)+g(x+ct)$,
$$u_{xx}=f''+g'',\qquad u_{tt}=c^2(f''+g'').$$
Thus $\boxed{u_{xx}=c^{-2}u_{tt}}$.


### Problem 24. Cancellation Identity
For $z=\cos(xy)$,
$$z_x=-y\sin(xy),\qquad z_y=-x\sin(xy).$$
Hence $\boxed{xz_x-yz_y=0}$.


### Problem 25. Reciprocal and Cyclic Relations
From $df=f_xdx+f_y dy+f_z dz=0$,
$$\boxed{(\partial y/\partial x)_z=-f_x/f_y,\quad (\partial x/\partial y)_z=-f_y/f_x,\quad (\partial y/\partial z)_x=-f_z/f_y,\quad (\partial z/\partial x)_y=-f_x/f_z.}
$$
These immediately give the reciprocal product $1$ and the cyclic product $-1$.


### Problem 26. Two Constraints in Three Variables
From $df=0$ and $dg=0$, with $z=z(x)$ and $y=y(x)$, solve the two linear equations for $dy/dx$:
$$\boxed{\frac{dy}{dx}=\frac{f_zg_x-f_xg_z}{f_yg_z-f_zg_y}.}$$


### Problem 27. Chain Rule with a Held Variable
If $u=u(x,y)$ and $y=y(x,z)$, then at fixed $z$,
$$\boxed{(\partial u/\partial x)_z=(\partial u/\partial x)_y+(\partial u/\partial y)_x(\partial y/\partial x)_z.}$$


### Problem 28. Specific-Heat Identity
Since $s=s(v,T)$ and $v=v(p,T)$,
$$
(\partial s/\partial T)_p=(\partial s/\partial T)_v+(\partial s/\partial v)_T(\partial v/\partial T)_p.
$$
Multiplying by $T$ gives
$$\boxed{c_p-c_v=T(s_v)_T(v_T)_p.}$$
