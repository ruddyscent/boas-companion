# 4-5. Chain Rule or Differentiating a Function of a Function


### Problem 1. One-Parameter Chain Rule
For $z=xe^{-y}$, $x=\cosh t$, $y=\cos t$,
$$
\boxed{\frac{dz}{dt}=e^{-y}\sinh t+xe^{-y}\sin t=e^{-\cos t}(\sinh t+\cosh t\sin t).}
$$


### Problem 2. Constant Radius in the $(u,v)$ Plane
Here $w=\sqrt{u^2+v^2}$ and $u=\cos A$, $v=\sin A$. Hence $w=1$ and
$$\boxed{dw/dp=0.}$$


### Problem 3. Exponential Composite
For $r=e^{-p^2-q^2}$, $p=e^s$, $q=e^{-s}$,
$$
\frac{dr}{ds}=r(-2pp'-2qq')=e^{-p^2-q^2}(-2p^2+2q^2).
$$
Thus
$$\boxed{dr/ds=2(e^{-2s}-e^{2s})e^{-e^{2s}-e^{-2s}}.}$$


### Problem 4. Logarithm Composite
For $x=\ln(u^2-v^2)$, $u=t^2$, $v=\cos t$,
$$
\boxed{\frac{dx}{dt}=\frac{2uu'-2vv'}{u^2-v^2}=\frac{4t^3+2\cos t\sin t}{t^4-\cos^2t}.}
$$


### Problem 5. Reduced Chain Rule
If $z=z(x,y)$ and $y=y(x)$, then
$$
dz=z_xdx+z_y dy=(z_x+z_yy')dx,
$$
so
$$\boxed{dz/dx=z_x+z_y\,dy/dx.}$$


### Problem 6. Power Composite
For $z=(x+y)^5$ and $y=\sin 10x$,
$$
\boxed{\frac{dz}{dx}=5(x+y)^4(1+10\cos 10x).}
$$


### Problem 7. Sine Composite
For $c=\sin(a-b)$ and $b=ae^{2a}$,
$$
\boxed{\frac{dc}{da}=\cos(a-b)\{1-e^{2a}(1+2a)\}.}
$$


### Problem 8. Differential Approximation with Dependent Variables
Let $x=x(t)$ and $y=y(t)$. From the total-change formula,
$$
\Delta z=z_x\Delta x+z_y\Delta y+\epsilon_1\Delta x+\epsilon_2\Delta y.
$$
Since $\Delta x=(x'+\eta_x)dt$ and $\Delta y=(y'+\eta_y)dt$, where $\eta_x,\eta_y\to0$,
$$
\Delta z=z_xdx+z_y dy+(\text{terms tending to }0)dt.
$$
Therefore $dz=z_xdx+z_y dy$ remains the principal part even when $x$ and $y$ are related through $t$.
