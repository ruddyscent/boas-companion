# 4-8. Application of Partial Differentiation to Maximum and Minimum Problems


### Problem 1. One-Variable Second Derivative Test
Taylor's formula at a stationary point gives $f(a+h)-f(a)=f''(a)h^2/2+O(h^3)$. For sufficiently small nonzero $h$, the sign is the sign of $f''(a)$. Thus $f''(a)>0$ gives a local minimum and $f''(a)<0$ gives a local maximum.


### Problem 2. Two-Variable Second Derivative Test
At a critical point,
$$f(a+h,b+k)-f(a,b)=\frac12(Ah^2+2Bhk+Ck^2)+O(\|(h,k)\|^3).$$
Complete the square:
$$Ah^2+2Bhk+Ck^2=A(h+Bk/A)^2+(C-B^2/A)k^2.$$
This proves the stated minimum, maximum, and saddle criteria according to the signs of $A,C$ and $AC-B^2$.


### Problem 3. Positive Quadratic
Complete squares:
$$x^2+y^2+2x-4y+10=(x+1)^2+(y-2)^2+5.$$
Thus $\boxed{(-1,2)}$ is a minimum point, value $5$.


### Problem 4. Saddle Quadratic
$$x^2-y^2+2x-4y+10=(x+1)^2-(y+2)^2+13.$$
The critical point $\boxed{(-1,-2)}$ is neither max nor min.


### Problem 5. Negative Definite Quadratic
The first derivatives give $1-2x-y=0$ and $1-x-y=0$, hence $(x,y)=(0,1)$. The Hessian has $f_{xx}=-2$, $f_{yy}=-1$, $f_{xy}=-1$, determinant $1>0$, so $\boxed{(0,1)}$ is a maximum.


### Problem 6. Cubic Saddle/Extremum Check
For $f=x^3-y^3-2xy+2$, $f_x=3x^2-2y$, $f_y=-3y^2-2x$. The critical points are $(0,0)$ and $(-2/3,2/3)$. The Hessian determinant is $-4$ at $(0,0)$, so that point is a saddle. At $(-2/3,2/3)$ the Hessian has negative diagonal entries and positive determinant, so it is a local maximum.


### Problem 7. A Point That Is Not an Extremum
For $z=(y-x^2)(y-2x^2)$, along $y=mx$ the leading term is $m^2x^2\ge0$, so every nonvertical line suggests a minimum. But along $y=(3/2)x^2$, $z=-x^4/4<0$, while along $y=0$, $z=2x^4>0$. Hence $(0,0)$ is neither a maximum nor a minimum.


### Problem 8. Roof Gutter
Let each side flap have length $x$ and bending angle $\theta$. The bottom width is $24-2x$ and the height is $x\sin\theta$, while the top width is $24-2x+2x\cos\theta$. The cross-section area is
$$A=x\sin\theta(24-2x+x\cos\theta).$$
Solving $A_x=A_\theta=0$ gives $\boxed{\theta=60^\circ,\\ x=8\text{ cm}}$, so the bottom is $8$ cm and each side is $8$ cm.


### Problem 9. Open Aquarium
For volume $V=xyz$ and material $S=xy+2xz+2yz$, eliminate $z=V/(xy)$. Then
$$S=xy+2V/y+2V/x.$$
The equations $S_x=S_y=0$ give $x=y$ and $x^3=2V$. Thus the base is square and $\boxed{z=x/2}$.


### Problem 10. Thicker Bottom Aquarium
If the bottom costs three times as much, minimize $S=3xy+2xz+2yz$ with $xyz=V$. This gives $x=y$ and $3x=2z$, so $\boxed{z=3x/2}$.


### Problem 11. Economical Tent
Write the area and volume in terms of the half-width, height, and length, then eliminate length using the fixed volume. Setting the two partial derivatives to zero gives the economical cross-section: the sloping sides meet the ground at $60^\circ$.


### Problem 12. Distance to $z=xy+5$
Minimize $D^2=x^2+y^2+(xy+5)^2$. The equations are
$$x+y(xy+5)=0,
\qquad y+x(xy+5)=0.$$
Subtracting gives $(x-y)(1-xy-5)=0$. If $x=y$, the point is $(0,0,5)$, with distance $5$. The other case is $xy+5=1$, so $xy=-4$; the first derivative equations then give $x+y=0$. Hence $(x,y)=(2,-2)$ or $(-2,2)$ and $z=1$. These give the smaller distance $\sqrt{4+4+1}=3$, so the shortest distance is $\boxed{3}$.


### Problem 13. Moment of Inertia for Three Masses
For $P=(x,y)$, minimize
$$I=m[x^2+(y-1)^2]+2m[(x-1)^2+y^2]+3m[(x-2)^2+(y-3)^2].$$
The minimum occurs at the mass-weighted average:
$$\boxed{P=\left(\frac{0+2+6}{6},\frac{1+0+9}{6}\right)=\left(\frac43,\frac53\right).}$$


### Problem 14. General Center of Mass
The same calculation gives
$$\boxed{P=\left(\frac{\sum m_i x_i}{\sum m_i},\frac{\sum m_i y_i}{\sum m_i}\right).}$$
Thus the point minimizing total moment of inertia is the center of mass.


### Problem 15. Closest Points on Two Lines
Parameterize the first line by $A(s)=(1-s,s,0)$ and the second by $B(t)=(t,t,t)$. Minimize $|A(s)-B(t)|^2$. Solving the two normal equations gives $s=1/2$, $t=1/3$. Thus the closest points are
$$\boxed{(1/2,1/2,0)\quad\text{and}\quad(1/3,1/3,1/3).}$$


### Problem 16. Least-Squares Line for Three Points
For $S=\sum(y_n-mx_n-b)^2$, the normal equations are
$$m\sum x_n^2+b\sum x_n=\sum x_ny_n,
\qquad m\sum x_n+Nb=\sum y_n.$$
For $(-1,-2),(0,0),(1,3)$, $\sum x=0$, $\sum y=1$, $\sum x^2=2$, $\sum xy=5$, so
$$\boxed{m=5/2,
\quad b=1/3.}$$


### Problem 17. Least-Squares Lines
Use the normal equations from Problem 16.

(a) For $(1,0),(2,-1),(3,-8)$:
$$\boxed{y=-4x+\frac{14}{3}.}$$

(b) For $(-2,-6),(-1,-3),(0,0),(1,9/2),(2,7)$:
$$\boxed{y=\frac{33}{10}x+\frac12.}$$

(c) For $(-2,4),(-1,0),(0,-1),(1,-8),(2,-10)$:
$$\boxed{y=-\frac{37}{10}x-3.}$$
