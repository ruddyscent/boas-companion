# 4-3. Total Differentials


### Problem 1. Taylor-Series Derivation of the Total Differential Error
Let $h=\Delta x$ and $k=\Delta y$. Taylor's formula gives
$$
f(a+h,b+k)-f(a,b)=f_x(a,b)h+f_y(a,b)k+R,
$$
where every term in $R$ has total degree at least $2$ in $h,k$. Hence
$$
\Delta z=dz+R,
\qquad dz=f_x(a,b)\Delta x+f_y(a,b)\Delta y.
$$
Write $R=\epsilon_1\Delta x+\epsilon_2\Delta y$ by grouping terms with a factor $\Delta x$ into the first product and the remaining higher-order terms into the second. Since all terms of $R$ are higher order, $\epsilon_1,\epsilon_2\to0$ as $(\Delta x,\Delta y)\to(0,0)$. Therefore
$$
\boxed{\Delta z=dz+\epsilon_1\Delta x+\epsilon_2\Delta y,\qquad \epsilon_1,\epsilon_2\to0.}
$$
