# 4-11. Change of Variables


### Problem 1. Factored Second-Order Operator
With $s=y+2x$, $t=y+3x$,
$$\partial_x=2\partial_s+3\partial_t,
\quad \partial_y=\partial_s+\partial_t.
$$
Substitution reduces the PDE to a nonzero constant times $z_{st}=0$, so
$$\boxed{z=F(s)+G(t)=F(y+2x)+G(y+3x).}$$


### Problem 2. Second Factored Operator
With $u=5x-2y$, $v=2x+y$, substitute
$$\partial_x=5\partial_u+2\partial_v,
\quad \partial_y=-2\partial_u+\partial_v.
$$
The PDE reduces to $z_{uv}=0$, hence
$$\boxed{z=F(5x-2y)+G(2x+y).}$$


### Problem 3. Wave-Type Equation with Source
Let $x=u+v$, $y=u-v$. Then the operator becomes a constant multiple of $w_{uv}$. The equation reduces to $w_{uv}=1$, so
$$\boxed{w=uv+F(u)+G(v).}$$
Returning to $x,y$, $u=(x+y)/2$, $v=(x-y)/2$.


### Problem 4. Differential Verification of Chain Rule
Write $dF=F_rdr+F_\theta d\theta$ and substitute $dr=r_xdx+r_y dy$, $d\theta=\theta_xdx+\theta_y dy$. The coefficient of $dx$ is $F_x$, giving
$$\boxed{F_x=F_r r_x+F_\theta\theta_x,}$$
and similarly for the other variables.


### Problem 5. Solving for Cartesian Operators
From
$$F_r=\cos\theta F_x+\sin\theta F_y,
\quad F_\theta=-r\sin\theta F_x+r\cos\theta F_y,
$$
solve the linear system:
$$\boxed{F_x=\cos\theta F_r-\frac{\sin\theta}{r}F_\theta,
\quad F_y=\sin\theta F_r+\frac{\cos\theta}{r}F_\theta.}$$


### Problem 6. Euler Equation to Constant Coefficients
With $x=e^z$, $dy/dx=e^{-z}y_z$ and $d^2y/dx^2=e^{-2z}(y_{zz}-y_z)$. Hence
$$x^2y''+2xy'-5y=0\quad\Rightarrow\quad\boxed{y_{zz}+y_z-5y=0.}$$


### Problem 7. Legendre Equation Change
With $x=\cos\theta$, $dx/d\theta=-\sin\theta$. Then
$$y_x=-\frac{y_\theta}{\sin\theta},
\quad y_{xx}=\frac{y_{\theta\theta}-\cot\theta y_\theta}{\sin^2\theta}.
$$
Substitution gives
$$\boxed{y_{\theta\theta}+\cot\theta y_\theta+2y=0.}$$


### Problem 8. Bessel Equation Change
With $u=2\sqrt{x}$, $x=u^2/4$. Chain-rule substitution reduces the equation to
$$\boxed{u^2y_{uu}+uy_u+(u^2-4)y=0.}$$


### Problem 9. Laplacian in Log-Polar Variables
For $x=e^s\cos t$, $y=e^s\sin t$, this is polar form with $r=e^s$. Since
$$\nabla^2u=u_{rr}+r^{-1}u_r+r^{-2}u_{tt}=e^{-2s}(u_{ss}+u_{tt}),$$
we get the stated identity.


### Problem 10. Legendre Transformations from $du=Tds-pdv$
Choose new variables by subtracting the product of each old variable and its conjugate:
$$\boxed{f=u-Ts,
\quad df=-s\,dT-p\,dv,}$$
$$\boxed{h=u+pv,
\quad dh=T\,ds+v\,dp,}$$
$$\boxed{g=u-Ts+pv,
\quad dg=-s\,dT+v\,dp.}$$


### Problem 11. Hamiltonian Transformation
Given $dL=\dot p\,dq+p\,d\dot q$, take
$$\boxed{H=p\dot q-L.}$$
Then
$$dH=\dot q\,dp-p\,d\dot q-dL+p\,d\dot q=\dot q\,dp-\dot p\,dq.$$


### Problem 12. Maxwell Relation from $u$
Since $du=Tds-pdv$, equality of mixed partials for $u(s,v)$ gives
$$\boxed{(\partial T/\partial v)_s=-(\partial p/\partial s)_v.}$$


### Problem 13. Three More Maxwell Relations
From the transforms in Problem 10:
$$\boxed{(\partial s/\partial v)_T=(\partial p/\partial T)_v,}$$
$$\boxed{(\partial T/\partial p)_s=(\partial v/\partial s)_p,}$$
$$\boxed{(\partial s/\partial p)_T=-(\partial v/\partial T)_p.}$$
