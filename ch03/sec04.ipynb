# 3-4. Vectors


### Problem 1. Vector Addition Law
Draw $A$ and $B$ head-to-tail. The diagonal of the parallelogram has the same initial and terminal points, so the geometric and component forms of vector addition agree.


### Problem 2. Diagrammatic Vector Algebra
Convert each vector to components, scale or subtract componentwise, and then draw the resulting arrows. The component identities follow because $(A+B)_x=A_x+B_x$ and similarly for the other displayed combinations.


### Problem 3. Parallelogram Diagonals
Let adjacent sides be $A$ and $B$. The midpoint of one diagonal is $(A+B)/2$, and the midpoint of the other is $A+(B-A)/2=(A+B)/2$; hence the diagonals bisect each other.


### Problem 4. Midpoint Segment in a Triangle
With sides represented by vectors from one vertex, the segment joining two midpoints is $\frac12B-\frac12A=\frac12(B-A)$, parallel to the third side and half as long.


### Problem 5. Trisection in a Parallelogram
Write the two midpoint lines parametrically and solve for their intersections with the diagonal. The parameters are $1/3$ and $2/3$, so the diagonal is trisected.


### Problem 6. Midpoints in a Quadrilateral
If three consecutive sides are $A,B,C$, the fourth is $-(A+B+C)$. The midpoint-joining vectors both have midpoint $\frac12(A+B+C)$ after translating from a common vertex, so the two joining segments bisect each other.


### Problem 7. Triangle Bisection by a Parallel Line
If the line through the midpoint is parallel to the second side, its direction is a scalar multiple of that side. Solving the vector equation for its intersection with the third side gives parameter $1/2$.


### Problem 8. Median of a Trapezoid
Let the parallel bases be $A$ and $B$. The segment joining the midpoints of the nonparallel sides has vector $\frac12(A+B)$; this proves both parallelism and length. The same midpoint calculation shows it bisects both diagonals.


### Problem 9. Component Combinations
With $A=(2,3)$ and $B=(4,-4)$, the requested vectors are $(-2,-3)$, $(12,-12)$, $(-2,7)$, $(8,2)$, and $(3,-\frac12)$.


### Problem 10. Recovering Two Vectors from Sum and Difference
Use $A=\frac12[(A+B)+(A-B)]$ and $B=\frac12[(A+B)-(A-B)]$. Thus $A=(0,\frac72)$ and $B=(-1,\frac12)$.


### Problem 11. Triangle Side Vectors
The directed sides are $AB=(-3,8,-6)$, $BC=(1,-10,3)$, and $CA=(2,2,3)$. Their sum is $0$, as it must be around a closed triangle.


### Problem 12. Angle Between Two Vectors
$A\cdot B=-6$, $|A|=3$, and $|B|=2\sqrt2$, so $\cos\theta=-1/\sqrt2$ and $\boxed{\theta=135^\circ}$.


### Problem 13. Scalar Projections
$A\cdot B=-5$, $|A|=5$, and $|B|=3$. The scalar projection of $A$ on $B$ is $-5/3$, that of $B$ on $A$ is $-1$, and $\cos\theta=-1/3$.


### Problem 14. Angles in a Cube
Two body diagonals can have dot product $1$ or $-1$ in units where each has length $\sqrt3$, so the acute angle has $\cos\theta=1/3$. A body diagonal and an edge have $\cos\theta=1/\sqrt3$. A body diagonal and a face diagonal have $\cos\theta=\sqrt{2/3}$.


### Problem 15. Unit and Perpendicular Vectors
For $A=(2,-1,2)$, $|A|=3$. A unit vector in its direction is $\frac13(2,-1,2)$; a length-$12$ vector is $4A=(8,-4,8)$. One perpendicular vector is $(1,2,0)$, and its unit version is $(1,2,0)/\sqrt5$.


### Problem 16. Angle Bisector by Unit Vectors
The unit vectors are $(2,-1,2)/3$ and $(-4,0,3)/5$. Their sum has equal dot product with each unit vector, so it lies along the angle bisector.


### Problem 17. A Vector Right Triangle
One example is $A=(1,1,1)$ and $B=(1,-1,0)$; since $A\cdot B=0$, the three sides $A$, $B$, and $A+B=(2,0,1)$ form a right triangle.


### Problem 18. Orthogonal Vectors and a Normal
The dot product is $0$. A vector perpendicular to both is their cross product, $\boxed{(-6,24,9)}$, or any nonzero scalar multiple.


### Problem 19. Common Perpendicular Vector
The cross product of the two given vectors is $(14,14,14)$, so $(1,1,1)$ is a convenient answer.


### Problem 20. Another Common Perpendicular
The cross product is $\boxed{(-2,2,-1)}$.


### Problem 21. Orthogonal Sum and Difference
Let $a=|A|$ and $b=|B|$. The dot product $(bA+aB)\cdot(bA-aB)=b^2A^2-a^2B^2=0$, so the two vectors are orthogonal.


### Problem 22. Law of Cosines
Squaring gives $(A+B)^2=A^2+B^2+2A\cdot B=A^2+B^2+2AB\cos\theta$, the vector form of the law of cosines.


### Problem 23. Zero Dot and Cross Conditions
$A\cdot B=0$ does not force $B=0$; take any nonzero vector perpendicular to $A$. $A\times B=0$ only forces $B$ parallel to $A$. If both hold and $A\ne0$, then $B$ is both parallel and perpendicular to $A$, so $B=0$.


### Problem 24. Lagrange Identity Special Case
Since $|A\times B|=AB\sin\theta$ and $A\cdot B=AB\cos\theta$, the sum is $\boxed{A^2B^2}$.


### Problem 25. Parallelogram Diagonal Squares
The diagonals are $A+B$ and $A-B$. Adding their squared lengths gives $2A^2+2B^2$.


### Problem 26. Median in an Isosceles Triangle
If the equal sides from the apex are $A$ and $B$ with $|A|=|B|$, then the median direction is $(A+B)/2$ and the base direction is $A-B$. Their dot product is $(A^2-B^2)/2=0$.


### Problem 27. Kite Diagonals
Write the equal adjacent sides as $A,B$ on one side and $C,D$ on the other with matching lengths. The diagonal directions reduce to a sum and difference of equal-length vectors, whose dot product is zero.


### Problem 28. Rhombus Diagonals
The diagonals are $A+B$ and $A-B$ for adjacent sides $A,B$ with $|A|=|B|$. Their dot product is $A^2-B^2=0$, and Problem 3 gives bisection.
