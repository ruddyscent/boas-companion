# 3-2. Matrices; Row Reduction


### Problem 1. Index Form of the Three Equations
Writing out the remaining two indexed equations gives
$$
M_{21}x_1+M_{22}x_2+M_{23}x_3=k_2,\qquad
M_{31}x_1+M_{32}x_2+M_{33}x_3=k_3.
$$
Substituting the entries of the coefficient matrix and putting
$(x_1,x_2,x_3)=(x,y,z)$ reproduces the second and third equations of the displayed system.


### Problem 2. Rectangular Systems in Index Notation
For two equations in four unknowns,
$$
\sum_{j=1}^4 M_{ij}x_j=k_i,\qquad i=1,2.
$$
For four equations in two unknowns,
$$
\sum_{j=1}^2 M_{ij}x_j=k_i,\qquad i=1,2,3,4.
$$
The row index counts equations and the column index counts unknowns.


### Problem 3. Unique Solution
$x=-3,\ y=5$; the coefficient and augmented ranks are both $2$.


### Problem 4. One-Parameter Family
Row reduction gives $x-\frac12z=\frac12,\ y=1$; hence $x=\frac12+\frac12t,\ y=1,\ z=t$.


### Problem 5. Inconsistent Pair
The reduced augmented matrix contains a row $0=1$, so there is no solution.


### Problem 6. One Free Variable
Reduction gives $x=1,\ y=z$. Thus $(x,y,z)=(1,t,t)$.


### Problem 7. Consistent Overdetermined System
The three equations reduce to $x=-4,\ y=3$; the extra equation is dependent after reduction.


### Problem 8. Dependent Equations
The reduced system is $x-y=-11,\ z=7$, so $(x,y,z)=(t-11,t,7)$.


### Problem 9. Inconsistent Three-Equation System
Reduction gives a row $0=1$, so the planes have no common point.


### Problem 10. Another Inconsistent System
Again the augmented matrix contains a row $0=1$; no solution exists.


### Problem 11. Three Independent Equations
The unique solution is $\boxed{(x,y,z)=(2,-1,-3)}$.


### Problem 12. Unique Solution
The unique solution is $\boxed{(x,y,z)=(-2,1,1)}$.


### Problem 13. One Free Variable
Reduction gives $x=-2,\ y=2z+\frac52$. Hence $(x,y,z)=(-2,2t+\frac52,t)$.


### Problem 14. Inconsistent Rank Test
The augmented rank exceeds the coefficient rank, so there is no solution.


### Problem 15. Rank of a 3 by 3 Matrix
The row-reduced form has two pivots, so $\boxed{\operatorname{rank}=2}$.


### Problem 16. Rank of a 3 by 4 Matrix
Three pivot rows occur; hence $\boxed{\operatorname{rank}=3}$.


### Problem 17. Full Rank Matrix
The row reduction gives the $4$ by $4$ identity, so $\boxed{\operatorname{rank}=4}$.


### Problem 18. Rank with One Dependent Row
There are three pivots and one zero row; therefore $\boxed{\operatorname{rank}=3}$.
