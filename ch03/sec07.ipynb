# 3-7. Linear Combinations, Linear Functions, Linear Operators


### Problem 1. Affine Scalar Function
Not linear: $f(0)=3\ne0$.


### Problem 2. Dot Product with a Projection
Linear: $A\cdot(r-kz)$ is a fixed linear combination of the components of $r$.


### Problem 3. Squared Norm
Not linear because $(cr)\cdot(cr)=c^2(r\cdot r)$ rather than $c(r\cdot r)$.


### Problem 4. Projection onto the yz-plane
Linear: $F(r)=r-ix$ removes the $x$ component and preserves sums and scalar multiples.


### Problem 5. Cross Product with a Fixed Vector
Linear in $r$ because $A\times(r+s)=A\times r+A\times s$ and $A\times(cr)=c(A\times r)$.


### Problem 6. Translation by a Fixed Vector
Not linear unless $A=0$, since $F(0)=A$.


### Problem 7. Definite Integral
Linear: integration distributes over sums and constants.


### Problem 8. Logarithm
Not linear: $\log(xy)=\log x+\log y$, not $\log x+\log y$ as an additive input rule for $x+y$. Also $\log(cx)\ne c\log x$ generally.


### Problem 9. Squaring
Not linear because $(u+v)^2\ne u^2+v^2$ in general.


### Problem 10. Reciprocal
Not linear: $(u+v)^{-1}$ is not $u^{-1}+v^{-1}$.


### Problem 11. Absolute Value
Not complex-linear since $|iz|=|z|$ but $i|z|$ is generally imaginary.


### Problem 12. Derivative Operators
$D,D^2,D^3,\ldots$ are linear by the linearity of differentiation.


### Problem 13. Polynomial Differential Operators
Both $D^2+2D+1$ and $x^2D^2-2xD+7$ are linear combinations of linear operations, so both are linear.


### Problem 14. Maximum Operator
Not linear: $\max(f+g)$ is not generally $\max f+\max g$.


### Problem 15. Transpose
Linear: $(A+B)^T=A^T+B^T$ and $(cA)^T=cA^T$.


### Problem 16. Inverse
Not linear: $(A+B)^{-1}$ is not $A^{-1}+B^{-1}$, and singular matrices may appear.


### Problem 17. Determinant
Not linear as a map on matrices; for example $\det(cA)=c^n\det A$.


### Problem 18. Linearity of the Cross Product
Resolve $B$ and $C$ into components perpendicular and parallel to $A$. The parallel components contribute zero, while in the perpendicular plane rotation by $90^\circ$ and scaling by $|A|$ distributes over vector addition.


### Problem 19. Complex Multiplication as Rotation
Taking real and imaginary parts of $e^{i\theta}(x+iy)$ gives $X=x\cos\theta-y\sin\theta$ and $Y=x\sin\theta+y\cos\theta$.


### Problem 20. Rotated Axes
Dotting $r=ix+jy=i'x'+j'y'$ with $i'$ and $j'$ gives $x'=x\cos\theta+y\sin\theta$ and $y'=-x\sin\theta+y\cos\theta$.


### Problem 21. Details of the Reflection Example
The listed matrices are orthogonal; the rotations have determinant $1$ and reflections determinant $-1$. Multiplication gives the two noncommuting products in the text. Solving $Cr=r$ gives the reflection line $y=-\sqrt3x$; for $D$ it is $y=\sqrt3x$.

### Problem 27. Rotation by Three Quarters of a Turn
Determinant $1$; the rotation angle is $3\pi/4$.


### Problem 28. Rotation about the x-axis
The rotation matrix is $\begin{pmatrix}1&0&0\\0&\cos\theta&-\sin\theta\\0&\sin\theta&\cos\theta\end{pmatrix}$. Composing with reflection through the $yz$-plane changes the first diagonal entry to $-1$.


### Problem 29. Rotation about y with Reflection
Using the active convention, $R_y(\pi/2)=\begin{pmatrix}0&0&1\\0&1&0\\-1&0&0\end{pmatrix}$. Reflection through the $xz$-plane multiplies the middle row by $-1$.


### Problem 30. Noncommuting Three-Dimensional Rotations
Multiplying the displayed $G$ and $K$ gives distinct matrices $GK$ and $KG$. Their fixed eigenvectors give different rotation axes, so the order of rotations matters.


### Problem 31. Book Rotation Experiment
The two products of $90^\circ$ rotations about different axes are distinct matrices. Each has trace $1$, so the rotation angle is $90^\circ$, but the fixed axes differ.


### Problem 32. Orthogonal Matrix Classification
