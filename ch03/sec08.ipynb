### Problem 16. Wronskian Criterion and Its Converse
If a linear relation existed, differentiating it $n-1$ times would give a homogeneous system for the coefficients. A nonzero Wronskian forces only the trivial coefficients. The functions $x^3$ and $|x^3|$ show that $W\equiv0$ need not imply dependence on an interval crossing $0$.
