### Problem 9. Odd Skew-Symmetric Example
The matrix satisfies $A^T=-A$. Hence
$$
\det A=\det A^T=\det(-A)=(-1)^3\det A=-\det A,
$$
so $\det A=0$.


### Problem 10. Odd-Order Skew-Symmetric Determinants
For an odd-order skew-symmetric matrix,
$$
\det A=\det A^T=\det(-A)=(-1)^n\det A=-\det A.
$$
Thus $\boxed{\det A=0}$.

### Problem 13. Three-Term Trigonometric Determinant
Expanding along the first row gives
$$
2\cos\theta(2\cos^2\theta-1)-2\cos\theta
=4\cos^3\theta-3\cos\theta=\boxed{\cos3\theta}.
$$


### Problem 14. Recurrence for the Tridiagonal Determinant
Let $D_n$ be the determinant. Expansion at the last row gives
$$
D_n=2\cos\theta\,D_{n-1}-D_{n-2}.
$$
With $D_0=1$ and $D_1=\cos\theta$, the addition formula for cosine gives $D_n=\cos n\theta$.

### Problem 16. Two-Unknown Quantum System
Cramer's rule gives
$$
A=\frac{k-iK}{k+iK}.
$$
Since the numerator is the conjugate of the denominator when $k,K$ are real,
$$
|A|^2=\frac{k^2+K^2}{k^2+K^2}=\boxed{1}.
$$


### Problem 17. Lorentz Equations
Putting the equations in standard form and applying Cramer's rule gives
$$
\boxed{x=\gamma(x'+vt')},\qquad
\boxed{t=\gamma\left(t'+\frac{v}{c^2}x'\right)}.
$$
The identity $\gamma^2(1-v^2/c^2)=1$ simplifies the determinants.


### Problem 18. Solving for the Third Unknown
The first two equations have the same $x-y$ coefficient structure. Subtracting them gives
$$
3b(x-y)-(3ab+6b^2)z=3b(b-a).
$$
Combining with the third equation and simplifying yields
$$
\boxed{z=\frac{b}{a+b}}
$$
when the displayed determinants are nonzero.
