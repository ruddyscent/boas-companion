# 3-13. A Brief Introduction to Groups


### Problem 1. Cyclic Rotations of Order Four
The matrices are $R(0),R(\pi/2),R(\pi),R(3\pi/2)$. Their products satisfy $R(a)R(b)=R(a+b)$ modulo $2\pi$, giving closure, identity, inverses, and associativity.


### Problem 2. Cyclic Groups of Orders Three and Six
Use powers of a generator $g$: $\{e,g,g^2\}$ with $g^3=e$, and $\{e,g,\ldots,g^5\}$ with $g^6=e$. The multiplication table is exponent addition modulo the order.


### Problem 3. Latin-Square Property
If $ab=ac$, left multiply by $a^{-1}$ to get $b=c$. Thus no row repeats an element; the column proof is analogous.


### Problem 4. The Four's Group
Multiplying the four matrices gives closure, identity $I$, and each nonidentity element squares to $I$. The group is Abelian but not cyclic because no element has order $4$.


### Problem 5. Abstract Group of Order Four
The table has $A^2=B^2=I$ and $AB=C$. It matches the Four's group, not the cyclic group, because all nonidentity elements have order $2$.


### Problem 6. Addition Modulo Four
The table is addition of exponents modulo $4$. The element $1$ generates the group, so it is isomorphic to the cyclic group of order $4$.


### Problem 7. Units Modulo Eight
The products of $1,3,5,7$ modulo $8$ stay in the set and each nonidentity element squares to $1$. Thus it is isomorphic to the Four's group.


### Problem 8. Symmetry Group of an Equilateral Triangle
Multiplying the rotation and reflection matrices in the stated order reproduces the table. Noncommuting products reflect the order of applying transformations.


### Problem 9. Cyclic Groups Are Abelian
Every element is a power of one generator: $g^ag^b=g^{a+b}=g^{b+a}=g^bg^a$.


### Problem 10. Symmetry Group of a Square
The group has four rotations and four reflections. The rotations are $R(k\pi/2)$ and the reflections are across the two axes and two diagonals; these eight matrices form the dihedral group $D_4$.


### Problem 11. Rectangle Symmetry Group
A non-square rectangle has identity, $180^\circ$ rotation, and two axis reflections. All nonidentity elements square to identity, so it is isomorphic to the Four's group.


### Problem 12. Classes in the Triangle Group
Conjugation groups the two nontrivial rotations together and the reflections together; $I$ is always alone because $C^{-1}IC=I$.


### Problem 13. Reducible and Irreducible Representations
The Abelian examples can be simultaneously diagonalized over the appropriate field, hence reducible. The triangle representation is not simultaneously diagonalizable because the matrices do not commute.


### Problem 14. Classes and Characters of the Square Group
The classes are $\{I\}$, $\{R^2\}$, $\{R,R^3\}$, the two axis reflections, and the two diagonal reflections. The character is the trace of the representative matrix on each class.


### Problem 15. Diagonalizing the Four's Group
The common eigenvectors $(1,1)$ and $(1,-1)$ form the columns of $C$. This diagonalizes all four matrices simultaneously.


### Problem 16. Diagonalizing the Cyclic Rotation Group
Over the complex numbers, use eigenvectors $(1,-i)$ and $(1,i)$ of the $90^\circ$ rotation. The diagonal entries are $1,i,-1,-i$ powers.


### Problem 17. Groups Listed in the Text
Verify closure, identity, inverses, and associativity for each listed set under its operation.


### Problem 18. Division Is Not a Group Operation
Associativity fails: $(a/b)/c$ is generally not $a/(b/c)$. Thus division cannot be the group law.


### Problem 19. Matrix Groups
Closure follows from determinant/product or orthogonality/unitarity identities; identity and inverses remain in the same set.


### Problem 20. Orthogonal Matrices with Determinant -1
Not a group: the product of two determinant $-1$ orthogonal matrices has determinant $+1$, so closure fails.


### Problem 21. SO(2) and SO(3)
$SO(2)$ is Abelian because planar rotations add angles. $SO(3)$ is not Abelian because rotations about different axes need not commute.
