# 3-14. General Vector Spaces


### Problem 1. Checking the Examples
For each example, test closure under addition and scalar multiplication, identify the zero vector, and check the inherited vector-space axioms.


### Problem 2. Span of Three Functions
This is a vector space by definition as a span. The listed functions are independent, so a basis is the listed set and the dimension is $3$.


### Problem 3. Function Span with Redundancy
The set is a vector space because it is a span. Remove dependent combinations; for example $(2-3e^x)\cos x$ and $x(1+5\cos x)$ are combinations of simpler listed terms where applicable after expanding.


### Problem 4. Polynomial Subspace
The condition $a_2=0$ is homogeneous, so this is a vector space. A basis is $\{1,x,x^3\}$ and the dimension is $3$.


### Problem 5. Polynomial Constraint
The condition $a_1=a_3$ is homogeneous. A basis is $\{1,x+x^3,x^2,x^4,x^5\}$, dimension $5$.


### Problem 6. Nonhomogeneous Polynomial Constraint
The condition $a_3=3$ excludes the zero polynomial and is not closed under scalar multiplication, so it is not a vector space.


### Problem 7. Even and Odd Coefficient Pattern
All even coefficients equal one parameter and all odd coefficients equal another. A basis is $\{1+x^2+x^4+x^6,\ x+x^3+x^5+x^7\}$, dimension $2$.


### Problem 8. Odd Powers Missing
This is the even-polynomial subspace through degree $6$. A basis is $\{1,x^2,x^4,x^6\}$, dimension $4$.


### Problem 9. Positive Coefficients
Positive even coefficients are not closed under multiplication by negative scalars and exclude zero, so this is not a vector space.


### Problem 10. Odd Coefficients Fixed by Even Coefficients
For each pair $(x^{2j},x^{2j+1})$, the basis element is $x^{2j}+\frac12x^{2j+1}$. Through degree $13$ this gives dimension $7$.


### Problem 11. Integral Inner Products
Linearity in the second slot, conjugate symmetry, and positivity follow directly from linearity of the integral and the fact that $\int |f|^2\ge0$.


### Problem 12. Deriving the Companion Inner-Product Rules
Take complex conjugates of the defining identities and use $\langle A|B\rangle=\langle B|A\rangle^*$ to get the relations in the first slot.


### Problem 13. Schwarz Inequality in Bracket Notation
Minimize $\langle A-\mu B|A-\mu B\rangle$ with $\mu=\langle B|A\rangle/\langle B|B\rangle$. Nonnegativity gives the inequality.


### Problem 14. Integral Form of Orthogonal Projection
Substituting $\langle f|g\rangle=\int f^*g$ into the projection formula gives exactly the displayed integral expression.


### Problem 15. Even-Odd Orthogonality
The product of an even power and an odd power is odd. Its integral over $(-1,1)$ is therefore $0$.


### Problem 16. Legendre-Type Orthogonalization Details
Apply Gram-Schmidt to $1,x,x^2,\ldots$. Odd terms are orthogonal to even terms by Problem 15; the omitted projection terms vanish for that reason. Normalizing by the computed integral norms gives the stated $e_i$.
