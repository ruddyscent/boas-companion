# 3-9. Special Matrices and Formulas


### Problem 1. Associativity in Index Notation
Both $((AB)C)_{ij}$ and $(A(BC))_{ij}$ expand to $\sum_k\sum_\ell a_{i\ell}b_{\ell k}c_{kj}$, so they are equal.


### Problem 2. Distributivity in Index Notation
$[A(B+C)]_{ij}=\sum_k a_{ik}(b_{kj}+c_{kj})=(AB)_{ij}+(AC)_{ij}$.


### Problem 3. Transpose, Inverse, Conjugate, Adjoint
Compute $A^T$, $\bar A$, and $A^\dagger=\bar A^T$ entrywise. The inverse is found by row reduction or cofactors, and direct multiplication verifies $AA^{-1}=I$.


### Problem 4. Second Complex Matrix
The same entrywise operations apply. The inverse check is the reliable final test: both $AA^{-1}$ and $A^{-1}A$ reduce to $I$.


### Problem 5. Symmetry of $AA^T$
$(AA^T)^T=(A^T)^TA^T=AA^T$, so the product is symmetric.


### Problem 6. Examples of Matrix Types
Examples are $\begin{pmatrix}1&2\\2&3\end{pmatrix}$, $\begin{pmatrix}0&2\\-2&0\end{pmatrix}$, any real matrix, and $i\begin{pmatrix}1&0\\0&-1\end{pmatrix}$.


### Problem 7. Index Notation Summary
Translate each formula by replacing matrix products with sums over repeated indices; for example $(AB)_{ij}=\sum_k a_{ik}b_{kj}$ and $(A^T)_{ij}=a_{ji}$.


### Problem 8. Adjoint of a Product
Using conjugation and transposition, $(AB)^\dagger=(\bar A\bar B)^T=\bar B^T\bar A^T=B^\dagger A^\dagger$. Induct on the number of factors for the general product rule.


### Problem 9. Two Meanings of Adjoint for Special Unitary Matrices
For a unitary matrix, $A^{-1}=A^\dagger$. If $\det A=1$, the cofactor adjoint also equals $A^{-1}$, so the two definitions coincide.


### Problem 10. Cofactors of a Proper Orthogonal Matrix
If $O^TO=I$ and $\det O=1$, then $O^{-1}=O^T$. Comparing with the cofactor formula for $O^{-1}$ shows each entry equals its own cofactor.


### Problem 11. Real Cases of Complex Notions
If $H=H^\dagger$ and entries are real, then $H=H^T$, so $H$ is symmetric. If $U^\dagger U=I$ and entries are real, then $U^TU=I$, so $U$ is orthogonal.


### Problem 12. Hermitian Entry Conditions
$A=A^\dagger$ means $a_{ij}=\bar a_{ji}$. In particular the diagonal entries equal their own conjugates, so they are real. An example is $\begin{pmatrix}2&1+i\\1-i&3\end{pmatrix}$.


### Problem 13. Unitary Matrix Check
Compute the conjugate transpose and multiply. The columns are orthonormal, so $U^\dagger U=I$ and the matrix is unitary.


### Problem 14. Transpose and Inverse Simplifications
$(AB^TC)^T=C^TBA^T$, $(C^{-1}MC)^{-1}=C^{-1}M^{-1}C$, and $(A^H)^{-1}(AHA^{-1})^3(HA^{-1})^{-1}$ simplifies by canceling adjacent inverse factors after preserving order.


### Problem 15. Pauli Matrix Identities
The Pauli matrices equal their adjoints, hence are Hermitian. Their commutators satisfy the cyclic relations, and expanding $[A,[B,C]]+[B,[C,A]]+[C,[A,B]]$ shows all terms cancel for any conformable matrices.


### Problem 16. Laplace Expansion with Cofactors
The sum $\sum_j a_{ij}C_{kj}$ is the determinant obtained by replacing row $k$ by row $i$. It is $\det A$ if $i=k$ and $0$ otherwise, giving $\delta_{ik}\det A$.


### Problem 17. Products Preserving Matrix Classes
For symmetric or Hermitian matrices, $(AB)^T=B^TA^T=BA$, so symmetry requires $AB=BA$. Products of orthogonal or unitary matrices remain in the same class by direct multiplication with their transpose or adjoint.


### Problem 18. Commutator of Symmetric Matrices
If $A^T=A$ and $B^T=B$, then $[A,B]^T=(AB-BA)^T=BA-AB=-[A,B]$, so the commutator is antisymmetric.


### Problem 19. Trace Identities
$\operatorname{Tr}(AB)=\sum_{ij}a_{ij}b_{ji}=\operatorname{Tr}(BA)$. Cyclic permutations preserve trace, but reversing order need not. If $S^T=S$ and $A^T=-A$, then $\operatorname{Tr}(SA)=-\operatorname{Tr}(SA)$, so it is $0$.


### Problem 20. Determinant of a Unitary Matrix
Taking determinants in $U^\dagger U=I$ gives $\overline{\det U}\det U=1$, hence $|\det U|=1$.


### Problem 21. Transpose of Sums and Powers
The sum rule follows entrywise. The power rule follows by induction from $(AB)^T=B^TA^T$ with all factors equal to $M$.


### Problem 22. Normal Matrices
Unitary matrices commute with their adjoints because $U^\dagger U=UU^\dagger=I$. The listed special classes similarly satisfy $AA^\dagger=A^\dagger A$ by their defining relations.


### Problem 23. Hermitian Combinations
$(AA^\dagger)^\dagger=AA^\dagger$, $(A+A^\dagger)^\dagger=A+A^\dagger$, and $[i(A-A^\dagger)]^\dagger=i(A-A^\dagger)$.


### Problem 24. Orthogonal Transformations Preserve Length
If $R=Mr$ and $M^TM=I$, then $R^TR=r^TM^TMr=r^Tr$, so $|R|=|r|$.


### Problem 25. Inverses and Similarity
If $O$ is orthogonal, $O^{-1}=O^T$, and the inverse also satisfies the orthogonality condition. The unitary case is identical with daggers. If $H=H^\dagger$ and $U$ is unitary, then $(U^{-1}HU)^\dagger=U^{-1}HU$.
