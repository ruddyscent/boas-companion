### Problem 1. Alternative Diagonalization Check
Substitute the stated eigenvectors into $Mr=\lambda r$. Reversing the eigenvector order changes $C$ by swapping columns and changes the order of the diagonal entries.


### Problem 2. Orthogonality of Eigenvectors
The dot product of the two normalized eigenvectors is $0$, and the columns of $C$ have unit length, so $C^TC=I$.


### Problem 3. Similarity Preserves Symmetry Type
If $C$ is orthogonal, $C^{-1}=C^T$. Then $(C^{-1}MC)^T=C^TM^TC$, which is the same matrix for symmetric $M$ and its negative for antisymmetric $M$.


### Problem 4. Inverse Rotation
The inverse of a rotation matrix is its transpose, obtained by replacing $\theta$ by $-\theta$.


### Problem 5. Rotation Angle in the Example

### Problem 6. Two Orthonormal Columns
If the two columns have unit length and zero dot product, then $C^TC=I$, so $C$ is orthogonal.


### Problem 7. n-Dimensional Version
The same argument says a matrix whose columns are an orthonormal basis satisfies $C^TC=I$ in any dimension.


### Problem 8. Lines Through the Origin Map to Lines
If $R=Mr$ and $M$ is invertible, then $r=M^{-1}R$. Substituting this into $y=mx$ gives a homogeneous linear equation in $X,Y$, hence another line through the origin.


### Problem 9. Determinant Under Similarity
$\det(C^{-1}MC)=\det(C^{-1})\det M\det C=\det M$. Thus the product of eigenvalues equals $\det M$.


### Problem 10. Trace Under Similarity
Using cyclic invariance of trace, $\operatorname{Tr}(C^{-1}MC)=\operatorname{Tr}(MCC^{-1})=\operatorname{Tr}M$. Thus the sum of eigenvalues equals the trace.
