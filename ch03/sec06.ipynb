# 3-6. Matrix Operations


### Problem 1. Basic Matrix Products
Compute each product by row-column multiplication. The products $AB$ and $BA$ differ, so matrix multiplication is not commutative. The determinant checks give
$$
\det(AB)=\det(BA)=\det A\,\det B,
$$
whereas sums and scalar multiples do not follow the naive scalar rules; for an $n$ by $n$ matrix, $\det(cA)=c^n\det A$.

### Problem 5. Products with Transposes
Both $AA^T$ and $A^TA$ are meaningful, but have different sizes unless $A$ is square. The same dimension check applies to $B$ and $C$; $CC^T=C^TC$ only if the actual products coincide, not merely because both exist.


### Problem 6. Pauli Matrix Products
Direct multiplication gives $A^2=B^2=C^2=I$, $AB=-BA=iC$, $BC=-CB=iA$, and $CA=-AC=iB$. Thus $[A,B]=2iC$ and cyclic permutations give the other commutators.

### Problem 9. Zero Product without Zero Factors

### Problem 10. Cancellation Failure
Although $AC=AD$, subtracting gives $A(C-D)=0$. Since $A$ is singular, cancellation is invalid; $C\ne D$.


### Problem 11. Identity Matrix

### Problem 23. Cofactor Formula Verification

### Problem 29. Exponentials of Noncommuting Matrices
In the series, the quadratic terms contain $AB+BA$ for $e^{A+B}$ but $2AB$ for $e^Ae^B$. These agree exactly when $AB=BA$; otherwise the exponentials need not be equal.


### Problem 30. Functions of a Pauli Matrix
Since $A^2=I$, even and odd powers separate: $\sin kA=A\sinh k$, $\cos kA=I\cosh k$, $e^{kA}=I\cosh k+A\sinh k$, and $e^{ikA}=I\cos k+iA\sin k$.


### Problem 31. Functions of a Diagonal Pauli Matrix
For $C=\operatorname{diag}(1,-1)$, apply the scalar function to diagonal entries. Thus $e^{kC}=\operatorname{diag}(e^k,e^{-k})$ and $e^{ikC}=\operatorname{diag}(e^{ik},e^{-ik})$.


### Problem 32. Rotation from a Pauli Matrix
Because $B^2=I$, $e^{i\theta B}=I\cos\theta+iB\sin\theta=\begin{pmatrix}\cos\theta&\sin\theta\\-\sin\theta&\cos\theta\end{pmatrix}$; replacing $\theta$ by $-\theta$ gives the inverse rotation.
