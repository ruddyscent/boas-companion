# 3-15. Miscellaneous Problems


### Problem 1. Linearity in One Determinant Row or Column
Expand the determinant along the row or column containing sums. Each cofactor is common to the two terms, so the determinant splits into a sum. Applying this to a scalar multiple of a row proves the corresponding determinant fact.


### Problem 2. Invalid Simultaneous Row Operation
The argument changes both rows using each other's original values at once. Elementary row replacement permits one row change at a time; doing both simultaneously is not determinant-preserving.

### Problem 5. Line and Plane Collection

### Problem 6. Point-Plane Distance Formula
Project the vector from a point on the plane to $(x_0,y_0,z_0)$ onto the normal $(a,b,c)$. This gives $|ax_0+by_0+cz_0-d|/\sqrt{a^2+b^2+c^2}$.


### Problem 7. Matrix Expression Meanings

### Problem 8. Complex Matrix Operations
Compute $A^T$, $\bar A$, and $A^\dagger$ entrywise. Row reduction gives $A^{-1}$; direct multiplication verifies the result.


### Problem 9. Thick Lens Matrix
Multiplying the three matrices gives a determinant $1$. The $(1,2)$ entry is $-\left[(n-1)(1/R_1-1/R_2)+d(n-1)^2/(nR_1R_2)\right]$, so this expression equals $1/f$.


### Problem 10. Two Thin Lenses
Multiplication gives $\det M=1$ and $1/f=1/f_1+1/f_2-d/(f_1f_2)$.


### Problem 11. Complex Product and Vector Products
If $z_1=x_1+iy_1$ and $z_2=x_2+iy_2$, then $z_1z_2^*=(x_1x_2+y_1y_2)+i(y_1x_2-x_1y_2)$. The real part is the dot product and the imaginary part is the signed planar cross product.


### Problem 12. Parallelogram Area Determinant

### Problem 31. Spring Frequencies
For constants $k,3k,k$ with equal masses, the characteristic values are proportional to $1$ and $7$, so $\omega=\sqrt{k/m}$ and $\sqrt{7k/m}$.


### Problem 32. Spring Frequencies

### Problem 33. Cayley-Hamilton after Diagonalization
If $D=C^{-1}MC$ is diagonal, then each diagonal entry satisfies the characteristic polynomial. Thus $p(D)=0$, and conjugating gives $p(M)=Cp(D)C^{-1}=0$.


### Problem 34. Noncommuting Exponentials
For Pauli matrices $A$ and $C$, $(A+C)^2=2I$, so $e^{A+C}=I\cosh\sqrt2+(A+C)\sinh\sqrt2/\sqrt2$. This differs from $e^Ae^C$ because $AC\ne CA$.


### Problem 35. Invertibility and Zero Eigenvalue
$A$ has an inverse iff $\det A\ne0$. The equation $Av=0$ has a nonzero solution exactly when $0$ is an eigenvalue, equivalently $\det A=0$.


### Problem 36. Half-Turn Rotation Group
The matrices are $\operatorname{diag}(1,-1,-1)$, $\operatorname{diag}(-1,1,-1)$, and $\operatorname{diag}(-1,-1,1)$, together with $I$. They commute and form the Four's group.


### Problem 37. Character of the Identity Class
The identity is represented by the $n$ by $n$ identity matrix in an $n$-dimensional irreducible representation, so its character is $\operatorname{Tr}I=n$.


### Problem 38. Classes in Abelian Groups
In a cyclic group, and more generally in any Abelian group, $g^{-1}ag=a$ for every $g$. Therefore every conjugacy class contains just one element.
