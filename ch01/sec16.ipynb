## Problems 1 and 2: Harmonic-Series Overhangs


<a id="problem-1"></a>

### Problem 1. Stacking Books

Let each book have length $L$, and count books from the top. Suppose a pile of $n-1$ books is placed on book $n$ at the point of tipping. The center of mass of the upper pile must lie directly over the right-hand edge of book $n$.

For book $n$, torque balance about its right-hand end gives

$$
(nW)d_n=\frac{WL}{2},
$$

where $d_n$ is the setback from the right end of book $n-1$ to the right end of book $n$. Therefore

$$
d_n=\frac{L}{2n}.
$$

The total overhang for $N$ books is

$$
D_N=\sum_{n=1}^N \frac{L}{2n}=\frac{L}{2}H_N.
$$

Since the harmonic series diverges, $D_N$ can be made larger than any prescribed multiple of $L$.

For 5 books,

$$
\frac{D_5}{L}=\frac12\left(1+\frac12+\frac13+\frac14+\frac15\right)
=1.1416\cdots,
$$

so the top book is completely to the right of the bottom book.

The number of books needed for overhangs of $2,3,10,100$ book lengths is approximately

$$
31,\qquad 227,\qquad 2.72\times10^8,\qquad 4.06\times10^{86}.
$$


```python
import math

def harmonic(n):
    return sum(1/k for k in range(1, n + 1))

for N in [25, 100, 200, 1000, 10**6]:
    print(f"H_{N} = {harmonic(N):.12f}")
print(f"H_(10^100) approx = {100 * math.log(10) + 0.5772156649015329:.12f}")

# Exact search for the small targets.
H = 0.0
found = {}
for n in range(1, 1000):
    H += 1/n
    for target in [2, 3]:
        if target not in found and H/2 > target:
            found[target] = n
print("books for overhang 2:", found[2])
print("books for overhang 3:", found[3])

for target in [10, 100]:
    estimate = math.exp(2 * target - 0.5772156649015329)
    print(f"books for overhang {target} approx: {estimate:.6e}")
```

### Problem 2. Mobile of Rods

Assume identical rods of length $L$ and weight $W$, numbered from the bottom. For rod $n$, the thread from above must support the total weight $nW$ of rod $n$ and everything below it.

Taking torques about the left end of rod $n$, the downward force of the rods below acts at the left end and contributes no torque. The rod's own weight acts at $L/2$, so

$$
(nW)x_n=W\frac{L}{2}.
$$

Thus the attachment point on rod $n$ should be

$$
\boxed{x_n=\frac{L}{2n}}
$$

from its left end. This is the same harmonic-series spacing as in Problem 1.


## Problems 3 through 8: Convergence Tests


<a id="problem-3"></a>

### Problem 3. The Series $\sum_{n=2}^{\infty}1/n^{3/2}$

This is a $p$-series with $p=3/2>1$, so it converges.

The proposed proof tries to compare the terms

$$
\frac1{\sqrt{2^3}},\frac1{\sqrt{3^3}},\frac1{\sqrt{4^3}},\ldots
$$

with

$$
\frac13,\frac16,\frac19,\frac1{12},\ldots .
$$

That comparison would require

$$
\sqrt{(n+1)^3}<3n
$$

for all large $n$. But this inequality is false eventually, since $(n+1)^{3/2}$ grows faster than $3n$. Thus the term-by-term lower comparison with a harmonic multiple fails. The correct test is simply the $p$-series test, which gives convergence.


<a id="problem-4"></a>

### Problem 4. $\sum_{n=1}^{\infty}2^n/n!$

Use the ratio test:

$$
\left|\frac{a_{n+1}}{a_n}\right|=\frac{2}{n+1}\to0.
$$

The series converges absolutely.


<a id="problem-5"></a>

### Problem 5. $\sum_{n=2}^{\infty}\frac{(n-1)^2}{1+n^2}$

The terms satisfy

$$
\frac{(n-1)^2}{1+n^2}\to 1.
$$

Since the terms do not tend to zero, the series diverges.


<a id="problem-6"></a>

### Problem 6. $\sum_{n=2}^{\infty}\frac{\sqrt{n-1}}{(n+1)^2-1}$

For large $n$,

$$
\frac{\sqrt{n-1}}{(n+1)^2-1}\sim \frac{\sqrt n}{n^2}=\frac{1}{n^{3/2}}.
$$

Since $\sum 1/n^{3/2}$ converges, the given series converges by limit comparison.


<a id="problem-7"></a>

### Problem 7. $\sum_{n=2}^{\infty}\frac{1}{n\ln(n^3)}$

Since

$$
\ln(n^3)=3\ln n,
$$

we have

$$
\sum_{n=2}^{\infty}\frac{1}{n\ln(n^3)}
=\frac13\sum_{n=2}^{\infty}\frac{1}{n\ln n}.
$$

This diverges by the integral test.


<a id="problem-8"></a>

### Problem 8. $\sum_{n=2}^{\infty}\frac{2n^3}{n^4-2}$

For large $n$,

$$
\frac{2n^3}{n^4-2}\sim \frac{2}{n}.
$$

The harmonic series diverges, so the given series diverges by limit comparison.


## Problems 9 through 13: Intervals of Convergence


<a id="problem-9"></a>

### Problem 9. $\sum_{n=1}^{\infty}\frac{x^n}{\ln(n+1)}$

The root test gives radius $R=1$. At $x=1$,

$$
\sum\frac{1}{\ln(n+1)}
$$

diverges, since its terms are eventually larger than $1/n$. At $x=-1$, the terms $1/\ln(n+1)$ decrease to zero, so the alternating series test gives convergence.

Thus the interval is

$$
\boxed{[-1,1)}.
$$


<a id="problem-10"></a>

### Problem 10. $\sum_{n=1}^{\infty}\frac{(n!)^2x^n}{(2n)!}$

For

$$
a_n=\frac{(n!)^2x^n}{(2n)!},
$$

we get

$$
\left|\frac{a_{n+1}}{a_n}\right|
=|x|\frac{(n+1)^2}{(2n+2)(2n+1)}\to \frac{|x|}{4}.
$$

So $R=4$. At $x=4$ or $x=-4$, the term magnitude is

$$
\frac{4^n(n!)^2}{(2n)!}=\frac{4^n}{\binom{2n}{n}}\sim \sqrt{\pi n},
$$

which does not tend to zero. Therefore both endpoints diverge, and the interval is

$$
\boxed{(-4,4)}.
$$


<a id="problem-11"></a>

### Problem 11. $\sum_{n=1}^{\infty}\frac{(-1)^n x^{2n-1}}{2n-1}$

This is $-\arctan x$, so the radius is $1$. At $x=1$ and $x=-1$, the series becomes an alternating reciprocal odd-integer series and converges conditionally.

Thus the interval is

$$
\boxed{[-1,1]}.
$$


<a id="problem-12"></a>

### Problem 12. $\sum_{n=1}^{\infty}\frac{x^n n^2}{5^n(n^2+1)}$

The coefficient $n^2/(n^2+1)$ tends to $1$, so the radius is $R=5$. At $x=5$, the terms tend to $1$; at $x=-5$, the terms do not tend to zero. Both endpoints diverge.

Therefore the interval is

$$
\boxed{(-5,5)}.
$$


<a id="problem-13"></a>

### Problem 13. $\sum_{n=1}^{\infty}\frac{(x+2)^n}{(-3)^n\sqrt n}$

This is a power series in $x+2$ with radius $3$. At $x=1$, the series is

$$
\sum_{n=1}^{\infty}\frac{(-1)^n}{\sqrt n},
$$

which converges by the alternating series test. At $x=-5$, the series is

$$
\sum_{n=1}^{\infty}\frac{1}{\sqrt n},
$$

which diverges.

Hence the interval is

$$
\boxed{(-5,1]}.
$$

## Problems 19 through 21: Taylor Series About a Point

### Problem 19. $\sin x$ About $a=\pi$

Let $h=x-\pi$. Then

$$
\sin x=\sin(\pi+h)=-\sin h.
$$

Thus

$$
\sin x=-(x-\pi)+\frac{(x-\pi)^3}{3!}-\frac{(x-\pi)^5}{5!}+\cdots.
$$

### Problem 20. $\sqrt[3]{x}$ About $a=8$

Let $h=x-8$. Then

$$
\sqrt[3]{x}=\sqrt[3]{8+h}=2\left(1+\frac{h}{8}\right)^{1/3}.
$$

Using the binomial series,

$$
\sqrt[3]{x}=2+\frac{x-8}{12}-\frac{(x-8)^2}{288}
+\frac{5(x-8)^3}{20736}+\cdots.
$$


<a id="problem-21"></a>

### Problem 21. $e^x$ About $a=1$

Let $h=x-1$. Then

$$
e^x=e^{1+h}=e\,e^h.
$$

Therefore

$$
e^x=e\left[1+(x-1)+\frac{(x-1)^2}{2!}
+\frac{(x-1)^3}{3!}+\cdots\right].
$$

$$
\sin\pi=\pi-\frac{\pi^3}{3!}+\frac{\pi^5}{5!}-\frac{\pi^7}{7!}+\cdots=0.
$$

Divide by $\pi$:

$$
1-\frac{\pi^2}{3!}+\frac{\pi^4}{5!}-\frac{\pi^6}{7!}+\cdots=0.

Rearranging,

$$
\frac{\pi^2}{3!}-\frac{\pi^4}{5!}+\frac{\pi^6}{7!}-\cdots=1.
$$

### Problem 24. $\ln3+\frac{(\ln3)^2}{2!}+\frac{(\ln3)^3}{3!}+\cdots=2$

Use the exponential series:

$$
e^u=1+u+\frac{u^2}{2!}+\frac{u^3}{3!}+\cdots.
$$

With $u=\ln3$,

$$
3=1+\ln3+\frac{(\ln3)^2}{2!}+\frac{(\ln3)^3}{3!}+\cdots.
$$

Subtract $1$ to get the result.


<a id="problem-25"></a>

### Problem 25. $\displaystyle\lim_{x\to0}\frac{x^2}{\ln\cos x}$

Since

$$
\cos x=1-\frac{x^2}{2}+\cdots,
$$

we have

$$
\ln\cos x=\ln\left(1-\frac{x^2}{2}+\cdots\right)
=-\frac{x^2}{2}+\cdots.
$$

Therefore

$$
\lim_{x\to0}\frac{x^2}{\ln\cos x}=-2.
$$

L'Hopital's rule gives the same result after differentiating twice.

\boxed{\frac23}.

Then

$$

## Problems 30 and 31: Integral Remainder Bounds

### Problem 30. Bounding $\zeta(1.1)$

For a positive decreasing sequence $a_n$, the tail after $N$ terms satisfies

$$
\int_{N+1}^{\infty}a(x)\,dx\le R_N\le \int_N^{\infty}a(x)\,dx.
$$

For $a_n=n^{-1.1}$,

$$
I_N=\int_N^{\infty}x^{-1.1}\,dx=10N^{-0.1}.
$$

To make the tail itself less than $0.005$, we need

$$
10N^{-0.1}<0.005,
$$

so

$$
N>2000^{10}=1.024\times10^{33}.
$$

For a much smaller computation, choose $N$ so that one term is about $0.005$:

$$
N^{-1.1}\approx0.005,
$$

which gives $N\approx124$. Then

$$
\sum_{n=1}^{124}\frac1{n^{1.1}}+10(125)^{-0.1}
\le \zeta(1.1)
\le
\sum_{n=1}^{124}\frac1{n^{1.1}}+10(124)^{-0.1}.
$$

Numerically this gives approximately

$$
10.58198\le \zeta(1.1)\le 10.58693.
$$

### Problem 31. More Integral Bounds

#### Part (a)

For

$$
\sum_{n=1}^{\infty}\frac1{n^{1.01}},
$$

the tail integral is

$$
I_N=\int_N^{\infty}x^{-1.01}\,dx=100N^{-0.01}.
$$

Tail error $<0.005$ would require

$$
N>20000^{100}\approx 10^{430.103}.
$$

Taking $N=190$ gives upper and lower bounds about

$$
100.57546\le S\le100.58044.
$$

#### Part (b)

For

$$
\sum_{n=1}^{\infty}\frac1{n(1+\ln n)^2},
$$

use

$$
I_N=\int_N^{\infty}\frac{dx}{x(1+\ln x)^2}=rac{1}{1+\ln N}.
$$

Tail error $<0.005$ requires $N>e^{199}$. Taking $N=15$ gives

$$
1.69054\le S\le1.69515.
$$

#### Part (c)

For

$$
\sum_{n=3}^{\infty}\frac1{n\ln n(\ln\ln n)^2},
$$

use

$$
I_N=\int_N^{\infty}\frac{dx}{x\ln x(\ln\ln x)^2}
=\frac{1}{\ln\ln N}.
$$

Tail error $<0.005$ requires $N>e^{e^{200}}$. Taking $N=35$ gives

$$
38.40437\le S\le38.40924.
$$


```python
import math

def bounds_p_series(p, N):
    partial = sum(n**(-p) for n in range(1, N + 1))
    upper = partial + (N**(1 - p))/(p - 1)
    lower = partial + ((N + 1)**(1 - p))/(p - 1)
    return lower, upper

lower, upper = bounds_p_series(1.1, 124)
print(f"Problem 30 bounds with N=124: {lower:.8f} <= zeta(1.1) <= {upper:.8f}")
print(f"Problem 30 terms for tail < 0.005: {(2000**10):.3e}")

lower, upper = bounds_p_series(1.01, 190)
print(f"31(a) bounds with N=190: {lower:.8f} <= S <= {upper:.8f}")
print(f"31(a) log10 terms for tail < 0.005: {100 * math.log10(20000):.3f}")

N = 15
partial = sum(1/(n*(1+math.log(n))**2) for n in range(1, N + 1))
lower = partial + 1/(1+math.log(N + 1))
upper = partial + 1/(1+math.log(N))
print(f"31(b) bounds with N=15: {lower:.8f} <= S <= {upper:.8f}")
print(f"31(b) log10 terms for tail < 0.005: {199 / math.log(10):.3f}")

N = 35
partial = sum(1/(n*math.log(n)*(math.log(math.log(n))**2)) for n in range(3, N + 1))
lower = partial + 1/math.log(math.log(N + 1))
upper = partial + 1/math.log(math.log(N))
print(f"31(c) bounds with N=35: {lower:.8f} <= S <= {upper:.8f}")
print(f"31(c) log10 of required N is about exp(200)/ln(10) = {math.exp(200)/math.log(10):.3e}")
```
