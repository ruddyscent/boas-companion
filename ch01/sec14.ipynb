### Problem 1. Error Bound for Alternating Series

Let

$$
S=a_1+a_2+a_3+\cdots
$$

be alternating, with $|a_{n+1}|<|a_n|$ and $a_n\to 0$. The error after $n$ terms is

$$
R_n=S-(a_1+a_2+\cdots+a_n)=a_{n+1}+a_{n+2}+a_{n+3}+\cdots.
$$

First group the tail as

$$
R_n=(a_{n+1}+a_{n+2})+(a_{n+3}+a_{n+4})+\cdots.
$$

Each pair has the sign of its first term, because the terms alternate and the magnitudes decrease. Hence $R_n$ has the same sign as $a_{n+1}$.

Now group the same tail as

$$
R_n=a_{n+1}+(a_{n+2}+a_{n+3})+(a_{n+4}+a_{n+5})+\cdots.
$$

Each parenthesized pair has the sign opposite to $a_{n+1}$. Therefore those pairs reduce the magnitude of the first term rather than increase it. Thus

$$
|R_n|\le |a_{n+1}|.
$$

So

$$
\left|S-(a_1+a_2+\cdots+a_n)\right|\le |a_{n+1}|.
$$

### Problem 2. Numerical Checks and Theorem (14.4)

#### Part (a)

The exact value is

$$
\sum_{n=1}^{\infty}\frac{1}{n^2}=\frac{\pi^2}{6}=1.6449\cdots.
$$

The sum of the first five terms is

$$
1+\frac14+\frac19+\frac1{16}+\frac1{25}=1.4636\cdots.
$$

Therefore the error is

$$
\frac{\pi^2}{6}-\left(1+\frac14+\frac19+\frac1{16}+\frac1{25}\right)
=0.1813\cdots.
$$

#### Part (b)

Using the known value

$$
\sum_{n=1}^{\infty}\frac{1}{n^2}\left(\frac12\right)^n
=\frac{\pi^2}{12}-\frac12(\ln 2)^2
=0.5822\cdots.
$$

The first five terms give

$$
\frac12+\frac1{16}+\frac1{72}+\frac1{256}+\frac1{800}
=0.5815\cdots.
$$

So the error is about

$$
0.5822405\cdots-0.5815451\cdots=0.000695\cdots.
$$


```python
import math

s_a = math.pi**2 / 6
partial_a = sum(1 / n**2 for n in range(1, 6))
error_a = s_a - partial_a

s_b = math.pi**2 / 12 - 0.5 * math.log(2)**2
partial_b = sum((1 / n**2) * (1 / 2)**n for n in range(1, 6))
error_b = s_b - partial_b

print(f"sum 1/n^2 = {s_a:.10f}")
print(f"first five = {partial_a:.10f}")
print(f"error = {error_a:.10f}")
print()
print(f"sum (1/n^2)(1/2)^n = {s_b:.10f}")
print(f"first five = {partial_b:.10f}")
print(f"error = {error_b:.10f}")
```

#### Part (c)

Let

$$
S=\sum_{n=0}^{\infty}a_nx^n,
$$

and suppose the series converges for $|x|<1$ and $|a_{n+1}|<|a_n|$ for $n>N$. The error after the terms through $n=N$ is

$$
R_N=\sum_{n=N+1}^{\infty}a_nx^n.
$$

By the triangle inequality,

$$
|R_N|\le \sum_{n=N+1}^{\infty}|a_n||x|^n.
$$

Since the coefficients are decreasing in absolute value after $N$,

$$
|a_n|\le |a_{N+1}|\qquad(n\ge N+1).
$$

Therefore

$$
|R_N|
\le |a_{N+1}|\sum_{n=N+1}^{\infty}|x|^n
=|a_{N+1}|\frac{|x|^{N+1}}{1-|x|}.
$$

Thus

$$
\boxed{|R_N|\le \frac{|a_{N+1}x^{N+1}|}{1-|x|}}.
$$


## Problems 3 through 8: Estimating Errors in Maclaurin Approximations

### Problem 3. Approximation for $\sqrt{1+x}$

The Maclaurin series is

$$
\sqrt{1+x}=1+\frac{x}{2}-\frac{x^2}{8}+\frac{x^3}{16}-\cdots.
$$

For $0<x<1/2$, the series is alternating after the first term and the term magnitudes decrease. If we approximate by

$$
\sqrt{1+x}\approx 1+\frac{x}{2},
$$

the first neglected term has magnitude $x^2/8$. Hence

$$
|R|\le \frac{x^2}{8}<\frac{(1/2)^2}{8}=\frac1{32}=0.03125<0.032.
$$

### Problem 4. Approximation for $\sin x$

Use

$$
\sin x=x-\frac{x^3}{3!}+\frac{x^5}{5!}-\cdots.
$$

Approximating by $\sin x\approx x$, the first neglected term is $x^3/3!$. By the alternating series estimate,

$$
|R|\le \frac{x^3}{6}.
$$

For $0<x<1/2$,

$$
|R|<\frac{(1/2)^3}{6}=\frac1{48}=0.02083\cdots<0.021.
$$

For $0<x<0.1$,

$$
|R|<\frac{(0.1)^3}{6}=0.000166\cdots<0.0002.
$$

### Problem 5. Approximation for $1-\cos x$

Since

$$
1-\cos x=\frac{x^2}{2!}-\frac{x^4}{4!}+\frac{x^6}{6!}-\cdots,
$$

approximating by $x^2/2$ leaves an alternating error whose first neglected term has magnitude $x^4/24$. Thus, for $|x|<1/2$,

$$
|R|\le \frac{|x|^4}{24}<\frac{(1/2)^4}{24}=\frac1{384}=0.002604\cdots<0.003.
$$

### Problem 6. Approximation for $\ln(1-x)$

Use

$$
\ln(1-x)=-\sum_{n=1}^{\infty}\frac{x^n}{n}
=-x-\frac{x^2}{2}-\frac{x^3}{3}-\cdots.
$$

If we approximate by $\ln(1-x)\approx -x$, the first neglected term is $-x^2/2$. The coefficients $1/n$ decrease, so theorem (14.4) gives

$$
|R|\le \frac{|x|^2/2}{1-|x|}.
$$

For $|x|<0.1$,

$$
|R|<\frac{0.1^2/2}{1-0.1}
=\frac{0.005}{0.9}=0.00555\cdots<0.0056.
$$

### Problem 7. Approximation for $2/\sqrt{4-x}$

Rewrite the function as

$$
\frac{2}{\sqrt{4-x}}=\frac{1}{\sqrt{1-x/4}}.
$$

Let $y=x/4$. Then $0<y<1/4$, and

$$
(1-y)^{-1/2}=1+\frac12y+\frac38y^2+\cdots.
$$

Since $y=x/4$,

$$
1+\frac12y=1+\frac{x}{8}.
$$

The first neglected term is $(3/8)y^2$. The positive coefficients decrease, so theorem (14.4) gives

$$
|R|\le \frac{(3/8)y^2}{1-y}.
$$

For $0<y<1/4$,

$$
|R|<\frac{(3/8)(1/4)^2}{1-1/4}
=\frac{3/128}{3/4}=\frac1{32}.
$$

Thus

$$
\frac{2}{\sqrt{4-x}}=1+\frac{x}{8}
$$

with error less than $1/32$ for $0<x<1$.

### Problem 8. Error Estimate for $\sum x^n/n^3$

Suppose

$$
S=\sum_{n=1}^{\infty}\frac{x^n}{n^3}
$$

is approximated by its first three terms. Then the first neglected term is

$$
\frac{x^4}{4^3}=\frac{x^4}{64}.
$$

The coefficients $1/n^3$ decrease, so theorem (14.4) gives

$$
|R_3|\le \frac{|x|^4/64}{1-|x|}.
$$

For $|x|<1/2$,

$$
|R_3|<\frac{(1/2)^4/64}{1-1/2}
=\frac{1}{512}=0.001953125.
$$

So the error is less than about $0.002$.


## Problems 9 and 10: When the First Neglected Term Is Not Enough


<a id="problem-9"></a>

### Problem 9. Remainder for the Series in Problem 4.6

Problem 4.6 is

$$
\sum_{n=1}^{\infty}\frac{1}{n(n+1)}.
$$

Since

$$
\frac{1}{n(n+1)}=\frac1n-\frac1{n+1},
$$

the partial sum is

$$
S_n=1-\frac1{n+1}.
$$

The full sum is $S=1$, so

$$
R_n=S-S_n=\frac1{n+1}.
$$

The first neglected term is

$$
a_{n+1}=\frac{1}{(n+1)(n+2)}.
$$

Thus

$$
\frac{R_n}{a_{n+1}}=n+2.
$$

So the error is not comparable to the first neglected term; it is larger by the factor $n+2$.


```python
for n in [3, 10, 100, 500]:
    remainder = 1 / (n + 1)
    next_term = 1 / ((n + 1) * (n + 2))
    print(f"n={n:3d}: R_n={remainder:.10f}, a_(n+1)={next_term:.10f}, ratio={remainder / next_term:.0f}")
```

### Problem 10. Convergence and Two-Decimal Accuracy

Consider

$$
\sum_{n=1}^{\infty}\frac{x^n}{n^2+n}
=\sum_{n=1}^{\infty}\frac{x^n}{n(n+1)}.
$$

By the ratio test,

$$
\lim_{n\to\infty}
\left|\frac{x^{n+1}/[(n+1)(n+2)]}{x^n/[n(n+1)]}\right|
=|x|.
$$

Thus the series converges for $|x|<1$ and diverges for $|x|>1$. At $x=1$,

$$
\sum_{n=1}^{\infty}\frac{1}{n(n+1)}
$$

converges by telescoping. At $x=-1$,

$$
\sum_{n=1}^{\infty}\left|\frac{(-1)^n}{n(n+1)}\right|
=\sum_{n=1}^{\infty}\frac{1}{n(n+1)}
$$

also converges. Hence the interval of convergence is

$$
\boxed{|x|\le 1}.
$$

For $x=1/2$, use theorem (14.4). After four terms, the first neglected term is

$$
\frac{(1/2)^5}{5^2+5}=\frac{1}{960}.
$$

Therefore

$$
|R_4|\le \frac{1/960}{1-1/2}=\frac1{480}=0.002083\cdots<0.005.
$$

Thus four terms give two-decimal-place accuracy.


## Problems 11 through 13: Convergence to the Function


<a id="problem-11"></a>

### Problem 11. Maclaurin Series for $\sin x$

For $f(x)=\sin x$, every derivative is one of

$$
\pm\sin x,\qquad \pm\cos x.
$$

Therefore

$$
|f^{(n+1)}(c)|\le 1
$$

for every real $c$. By the Taylor remainder formula,

$$
|R_n(x)|=
\left|\frac{x^{n+1}f^{(n+1)}(c)}{(n+1)!}\right|
\le \frac{|x|^{n+1}}{(n+1)!}.
$$

For fixed $x$,

$$
\lim_{n\to\infty}\frac{|x|^{n+1}}{(n+1)!}=0.
$$

Hence $R_n(x)\to 0$, so the Maclaurin series for $\sin x$ converges to $\sin x$.


<a id="problem-12"></a>

### Problem 12. Maclaurin Series for $e^x$

For $f(x)=e^x$, all derivatives are $e^x$. The Taylor remainder is

$$
R_n(x)=\frac{x^{n+1}e^c}{(n+1)!},
$$

where $c$ lies between $0$ and $x$.

For fixed $x$, $e^c\le e^{|x|}$ on the interval between $0$ and $x$. Therefore

$$
|R_n(x)|\le e^{|x|}\frac{|x|^{n+1}}{(n+1)!}.
$$

Since

$$
\lim_{n\to\infty}\frac{|x|^{n+1}}{(n+1)!}=0,
$$

we get $R_n(x)\to 0$. Thus the Maclaurin series for $e^x$ converges to $e^x$.

### Problem 13. Maclaurin Series for $(1+x)^p$ when $0<x<1$

Let

$$
f(x)=(1+x)^p.
$$

The $(n+1)$st derivative is

$$
f^{(n+1)}(x)=p(p-1)(p-2)\cdots(p-n)(1+x)^{p-n-1}.
$$

By Taylor's theorem about $0$,

$$
R_n(x)=\frac{x^{n+1}}{(n+1)!}
 p(p-1)\cdots(p-n)(1+c)^{p-n-1},
$$

where $c$ lies between $0$ and $x$.

For all sufficiently large $n$, the exponent $p-n-1$ is negative. Since $0<c<x<1$, we have $1+c>1$, and therefore

$$
(1+c)^{p-n-1}\le 1
$$

for those large $n$. Hence

$$
|R_n(x)|\le
\left|\frac{p(p-1)\cdots(p-n)}{(n+1)!}\right|x^{n+1}
=\left|\binom{p}{n+1}\right|x^{n+1}.
$$

Let

$$
b_n=\left|\binom{p}{n+1}\right|x^{n+1}.
$$

Then

$$
\frac{b_{n+1}}{b_n}
=\left|\frac{p-n-1}{n+2}\right|x\to x<1.
$$

Therefore $b_n\to 0$, and so $R_n(x)\to 0$. Thus the Maclaurin series for $(1+x)^p$ converges to $(1+x)^p$ when $0<x<1$.
