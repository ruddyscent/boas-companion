## Problems 1 through 4: Numerical Computation by Series


<a id="problem-1"></a>

### Problem 1. $e^{\arcsin x}+\ln\left(\frac{1-x}{e}\right)$ at $x=0.0003$

Use

$$
\arcsin x=x+\frac{x^3}{6}+\frac{3x^5}{40}+\cdots
$$

and expand $e^{\arcsin x}$. Combining it with

$$
\ln\left(\frac{1-x}{e}\right)=\ln(1-x)-1
$$

gives the cancellation

$$
e^{\arcsin x}+\ln\left(\frac{1-x}{e}\right)
=-\frac{x^4}{24}-\frac{x^5}{30}-\frac{7x^6}{144}-\cdots.
$$

Thus at $x=0.0003$,

$$
-\frac{x^4}{24}-\frac{x^5}{30}\approx -3.37581\times 10^{-16}.
$$

A direct floating-point evaluation loses digits because the leading terms cancel.


<a id="problem-2"></a>

### Problem 2. $\dfrac{1}{\sqrt{1+x^4}}-\cos x^2$ at $x=0.012$

Use

$$
(1+x^4)^{-1/2}=1-\frac{x^4}{2}+\frac{3x^8}{8}-\frac{5x^{12}}{16}+\cdots
$$

and

$$
\cos x^2=1-\frac{x^4}{2}+\frac{x^8}{24}-\frac{x^{12}}{720}+\cdots.
$$

Subtracting,

$$
\frac{1}{\sqrt{1+x^4}}-\cos x^2
=\frac{x^8}{3}-\frac{14x^{12}}{45}+\cdots.
$$

At $x=0.012$ this is approximately

$$
\frac{x^8}{3}=1.43327\times 10^{-16}.
$$


<a id="problem-3"></a>

### Problem 3. $\ln\left(x+\sqrt{1+x^2}\right)-\sin x$ at $x=0.001$

Since

$$
\ln\left(x+\sqrt{1+x^2}\right)=\operatorname{arsinh}x
=x-\frac{x^3}{6}+\frac{3x^5}{40}-\frac{5x^7}{112}+\cdots,
$$

and

$$
\sin x=x-\frac{x^3}{6}+\frac{x^5}{120}-\frac{x^7}{5040}+\cdots,
$$

we get

$$
\ln\left(x+\sqrt{1+x^2}\right)-\sin x
=\frac{x^5}{15}-\frac{2x^7}{45}+\cdots.
$$

At $x=0.001$,

$$
\frac{x^5}{15}\approx 6.66667\times 10^{-17}.
$$


<a id="problem-4"></a>

### Problem 4. $e^{\sin x}-x^{-3}\ln(1+x^3e^x)$ at $x=0.00035$

Expand both terms about $x=0$. The first nonzero terms of the difference are

$$
e^{\sin x}-\frac{1}{x^3}\ln(1+x^3e^x)
=\frac{x^3}{3}+\frac{5x^4}{6}+\frac{37x^5}{40}+\cdots.
$$

Thus at $x=0.00035$,

$$
\frac{x^3}{3}+\frac{5x^4}{6}+\frac{37x^5}{40}
\approx 1.43042\times 10^{-11}.
$$

Direct evaluation is unreliable here because $x^{-3}\ln(1+x^3e^x)$ subtracts two nearly equal quantities after a severe scaling by $x^{-3}$.


```python
import math

items = []
x = 0.0003
items.append(("1", -x**4/24 - x**5/30, math.exp(math.asin(x)) + math.log((1 - x) / math.e)))

x = 0.012
items.append(("2", x**8/3 - 14*x**12/45, 1 / math.sqrt(1 + x**4) - math.cos(x**2)))

x = 0.001
items.append(("3", x**5/15 - 2*x**7/45, math.log(x + math.sqrt(1 + x*x)) - math.sin(x)))

x = 0.00035
items.append(("4", x**3/3 + 5*x**4/6 + 37*x**5/40, math.exp(math.sin(x)) - x**-3 * math.log(1 + x**3 * math.exp(x))))

for label, series_value, direct_value in items:
    print(f"Problem {label}: series {series_value:.18e}, direct {direct_value:.18e}")
```

## Problems 14 and 15: Integral Approximations


<a id="problem-14"></a>

### Problem 14. $\displaystyle\int_0^t e^{-x^2}\,dx$, $0<t<0.1$

Use

$$
e^{-x^2}=1-x^2+\frac{x^4}{2!}-\frac{x^6}{3!}+\cdots.
$$

Integrating term by term,

$$
\int_0^t e^{-x^2}\,dx
=t-\frac{t^3}{3}+\frac{t^5}{10}-\cdots.
$$

A two-term approximation is

$$
\boxed{t-\frac{t^3}{3}}.
$$

The next term gives the alternating-series error bound

$$
|R|<\frac{t^5}{10}<\frac{(0.1)^5}{10}=10^{-6}.
$$


<a id="problem-15"></a>

### Problem 15. $\displaystyle\int_0^t \sqrt{x}e^{-x}\,dx$, $0<t<0.01$

Use

$$
\sqrt{x}e^{-x}=x^{1/2}\left(1-x+\frac{x^2}{2!}-\cdots\right).
$$

Then

$$
\int_0^t \sqrt{x}e^{-x}\,dx
=\frac{2}{3}t^{3/2}-\frac{2}{5}t^{5/2}+\frac{1}{7}t^{7/2}-\cdots.
$$

A two-term approximation is

$$
\boxed{\frac{2}{3}t^{3/2}-\frac{2}{5}t^{5/2}}.
$$

The next term gives

$$
|R|<\frac{t^{7/2}}{7}<\frac{(0.01)^{7/2}}{7}=\frac{10^{-7}}{7}.
$$

### Problem 25. Why Problem 24(f) Can Use the Series for $e^x$

Write

$$
x^ne^{-x}=\frac{x^n}{e^x}.
$$

The exponential series contains arbitrarily high powers:

$$
e^x=1+x+\frac{x^2}{2!}+\cdots.
$$

For any fixed $n$, choose an integer $m>n$. Then

$$
e^x\ge \frac{x^m}{m!}
$$

for $x>0$, so

$$
0\le \frac{x^n}{e^x}\le m!\,x^{n-m}\to0.
$$

The special feature of the $e^x$ series is that all its terms are positive for $x>0$ and the series contains powers higher than any fixed $n$.

### Problem 26. Derivatives of $e^{-1/t^2}$ at $t=0$

Define

$$
f(t)=\begin{cases}
e^{-1/t^2},&t\ne0,\\
0,&t=0.
\end{cases}
$$

For $t\ne0$,

$$
f'(t)=\frac{2}{t^3}e^{-1/t^2},
$$

and

$$
f''(t)=\left(\frac{4}{t^6}-\frac{6}{t^4}\right)e^{-1/t^2}.
$$

Each higher derivative has the form

$$
P\left(\frac1t\right)e^{-1/t^2},
$$

where $P$ is a polynomial. Put $x=1/t^2$. Then every term is a power of $x$ times $e^{-x}$, which tends to $0$ as $x\to\infty$. Therefore

$$
f(0)=f'(0)=f''(0)=\cdots=0.
$$

This is the standard example of a nonzero function whose Maclaurin series is identically zero.

### Problem 31. Infinite Tower of Circular Disks

The disk at height $n$ has radius

$$
r_n=\frac{1}{n\ln n},\qquad n\ge2.
$$

#### Part (a)

The total area is

$$
\sum_{n=2}^{\infty}\pi r_n^2
=\pi\sum_{n=2}^{\infty}\frac{1}{n^2(\ln n)^2}.
$$

Since

$$
\frac{1}{n^2(\ln n)^2}\le \frac{1}{n^2}
$$

for $n\ge2$, the total area is finite.

#### Part (b)

The total wire length is

$$
\sum_{n=2}^{\infty}2\pi r_n
=2\pi\sum_{n=2}^{\infty}\frac{1}{n\ln n}.
$$

This diverges by the integral test, so the required wire length is infinite.

#### Part (c)

There is no contradiction because area and length have different units. If each strip width is a fixed fraction of its disk radius, then the strip areas are comparable to $r_n^2$, so the total strip area is finite even though the circumferences have infinite total length. If the strips all have one fixed positive width, then eventually that width exceeds the radius of the small disks, so the construction cannot continue in the stated way.

### Problem 32. Doubling Time Rule

We want

$$
\left(1+\frac{i}{100}\right)^n=2.
$$

Taking logarithms gives

$$
n\ln\left(1+\frac{i}{100}\right)=\ln2.
$$

Let $r=i/100$. Then

$$
ni=100\ln2\,\frac{r}{\ln(1+r)}.
$$

Since $100\ln2=69.3147\cdots$, and for small $r$,

$$
\ln(1+r)=r-\frac{r^2}{2}+\frac{r^3}{3}-\cdots,
$$

we get the approximation

$$
\boxed{ni\approx 69}.
$$

For $i\le20$, $r\le0.2$, and

$$
\frac{r}{\ln(1+r)}\le \frac{0.2}{\ln1.2}=1.097\cdots.
$$

Thus the correction from using $69$ is less than about $10\%$ over this range.

### Problem 33. Distance to the Horizon

From the figure,

$$
\frac{h}{R}=\sec\theta-1.
$$

For small $\theta$,

$$
\sec\theta=1+\frac{\theta^2}{2}+\cdots.
$$

Thus

$$
\frac{h}{R}\approx \frac{\theta^2}{2},
\qquad
\theta\approx \sqrt{\frac{2h}{R}}.
$$

Since the surface distance is $s=R\theta$,

$$
s\approx R\sqrt{\frac{2h}{R}}=\sqrt{2Rh}.
$$

If $R\approx3960$ miles and $h$ is in feet, then

$$
s\text{ miles}
=\frac{\sqrt{2(3960)(5280)h}}{5280}
=\sqrt{\frac{3h}{2}}.
$$

So

$$
\boxed{s\approx \sqrt{\frac{3h}{2}}\text{ miles}}.
$$
