### Problem 6. $(-e)^{i\pi}$

### Problem 13. Real Variables from a Modulus Equation
Let $z=x+iy$. Since $|z+3|$ is real, $1-iz=1+y-ix$ must be real, so $x=0$. Then $|y i+3|=1+y$, giving $9+y^2=(1+y)^2$, hence $\boxed{x=0,
 y=4}$.


### Problem 14. Disk of Convergence
For $\sum (z-2i)^n/n$, the radius is $1$, so
$$\boxed{|z-2i|<1.}$$


### Problem 15. Absolute Convergence of $\sum z^{\ln n}$
Using $z^{\ln n}=e^{(\ln n)\ln z}=n^{\ln z}$, absolute convergence depends on $n^{\operatorname{Re}\ln z}$. Thus it converges when
$$\boxed{\operatorname{Re}\ln z<-1.}$$


### Problem 16. Half-Plane from a Rotated Product
Since $e^{i\pi/2}=i$, $\operatorname{Re}(iz)>2$. If $z=x+iy$, then $iz=-y+ix$, so the set is
$$\boxed{y<-2.}$$


### Problem 17. Formula for $\arcsin z$
Set $w=\arcsin z$, so $z=(e^{iw}-e^{-iw})/(2i)$. Solving the quadratic for $e^{iw}$ gives the logarithmic formula.


### Problem 18. Formula for $\arccos z$

### Problem 19. Formula for $\arctan z$
Set $z=\tan w$ and solve for $e^{2iw}$ to obtain the logarithm.


### Problem 20. Formula for $\sinh^{-1} z$

### Problem 21. Formula for $\cosh^{-1} z$

### Problem 22. Formula for $\tanh^{-1} z$

### Problem 25. Conjugation Properties
For real-coefficient power series, conjugation passes through each term, so $\overline{f(z)}=f(\bar z)$. This proves the stated cosine identity and explains which sine or linear examples also hold.


### Problem 26. A Modulus Depending on $\theta$
Compute the square of the modulus by multiplying by the conjugate:
$$\left|\frac{2e^{i\theta}-i}{ie^{i\theta}+2}\right|^2
=\frac{(2e^{i\theta}-i)(2e^{-i\theta}+i)}{(ie^{i\theta}+2)(-ie^{-i\theta}+2)}.$$
Simplification gives the desired real expression.

(b) If $z=x+iy$, then $|e^z|^2=e^{2x}=\boxed{e^{2\operatorname{Re}z}}$.

(c) Apply part (b) to the exponent and simplify the real part.


### Problem 28. Quantum-Mechanics Absolute Square

### Problem 29. Algebraic Equation for $z=a/b$
From $1/(a+b)=1/a+1/b=(a+b)/(ab)$, get $ab=(a+b)^2$. Dividing by $b^2$ gives
$$z=(z+1)^2,$$
so
$$\boxed{z=\frac{-1\pm i\sqrt3}{2}.}$$


### Problem 30. Series for $e^{x(1+i)}$
Since $1+i=\sqrt2e^{i\pi/4}$,
$$\frac{(1+i)^n}{n!}x^n=\frac{2^{n/2}e^{in\pi/4}}{n!}x^n.$$
Real parts give $e^x\cos x$ and imaginary parts give $e^x\sin x$. Terms vanish when the corresponding cosine or sine factor is zero.


### Problem 31. Sequence and Absolute Values
If $z_n=a_n+ib_n\to0$, then $|z_n|=\sqrt{a_n^2+b_n^2}\to0$. Conversely, $|a_n|,|b_n|\le |z_n|$, so $|z_n|\to0$ implies $z_n\to0$.


### Problem 32. Exponential Series
Using the exponential series,
$$\sum_{n=0}^\infty\frac{(1+i\pi)^n}{n!}=e^{1+i\pi}=e\cdot(-1)=\boxed{-e}.
$$
