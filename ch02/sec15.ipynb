# 2-15. Inverse Trigonometric and Hyperbolic Functions

Use the logarithmic formulas for inverse functions; all logarithms are multivalued unless a principal value is specified.


### Problem 1. $\arcsin 2$
$$\boxed{\arcsin2=-i\ln(2i\pm i\sqrt3).}$$


### Problem 2. $\arctan 2i$
$$\boxed{\arctan(2i)=\frac1{2i}\ln\frac{1-2}{1+2}=\frac1{2i}\ln(-1/3).}$$


### Problem 3. $\cosh^{-1}(1/2)$
$$\boxed{\cosh^{-1}(1/2)=\ln\left(\frac12\pm\frac{\sqrt3}{2}i\right).}$$


### Problem 4. $\sinh^{-1}(i/2)$
$$\boxed{\sinh^{-1}(i/2)=\ln\left(\frac i2\pm\frac{\sqrt3}{2}\right).}$$


### Problem 5. $\arccos(i\sqrt8)$
$$\boxed{\arccos(i\sqrt8)=i\ln\left(i\sqrt8\pm\sqrt{-9}\right).}$$


### Problem 6. $\tanh^{-1}(-i)$
The formula gives a logarithm with denominator $1+i$ and numerator $1-i$; this is a limiting branch value, reflecting that $\tanh z=-i$ is reached at poles/branch choices.


### Problem 7. $\arctan(i\sqrt2)$
$$\boxed{\arctan(i\sqrt2)=\frac1{2i}\ln\frac{1-\sqrt2}{1+\sqrt2}.}$$


### Problem 8. $\arcsin(5/3)$
$$\boxed{\arcsin(5/3)=-i\ln\left(\frac{5i}{3}\pm\frac{4i}{3}\right).}$$


### Problem 9. $\tanh^{-1}(i\sqrt3)$
$$\boxed{\tanh^{-1}(i\sqrt3)=\frac12\ln\frac{1+i\sqrt3}{1-i\sqrt3}.}$$


### Problem 10. $\arccos(5/4)$
$$\boxed{\arccos(5/4)=i\ln\left(\frac54\pm\frac34\right).}$$


### Problem 11. $\sinh^{-1}(i/\sqrt2)$
$$\boxed{\sinh^{-1}(i/\sqrt2)=\ln\left(\frac{i}{\sqrt2}\pm\frac1{\sqrt2}\right).}$$


### Problem 12. $\cosh^{-1}(\sqrt3/2)$
$$\boxed{\cosh^{-1}(\sqrt3/2)=\ln\left(\frac{\sqrt3}{2}\pm\frac i2\right).}$$


### Problem 13. $\cosh^{-1}(-1)$
$$\boxed{\cosh^{-1}(-1)=i(\pi+2\pi n).}$$


### Problem 14. $\arcsin(3i/4)$
$$\boxed{\arcsin(3i/4)=-i\ln\left(-\frac34\pm\frac54\right).}$$


### Problem 15. $\arctan(2+i)$
$$\boxed{\arctan(2+i)=\frac1{2i}\ln\frac{1+i(2+i)}{1-i(2+i)}.}$$


### Problem 16. $\tanh^{-1}(1-2i)$
$$\boxed{\tanh^{-1}(1-2i)=\frac12\ln\frac{2-2i}{2i}.}$$


### Problem 17. Tangent Omits $\pm i$
Solving $\tan z=i$ gives $(e^{iz}-e^{-iz})/(i(e^{iz}+e^{-iz}))=i$, which reduces to $e^{-iz}=0$, impossible. The case $-i$ similarly gives $e^{iz}=0$.


### Problem 18. Hyperbolic Tangent Omits $\pm1$
Solving $\tanh z=1$ gives $e^{-z}=0$, impossible, and solving $\tanh z=-1$ gives $e^z=0$, impossible.
