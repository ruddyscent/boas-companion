# 2-14. Complex Roots and Powers

For logarithms, use $\ln(re^{i\theta})=\ln r+i(\theta+2\pi n)$.


### Problem 1. $\ln(-e)$
$$\boxed{\ln(-e)=1+i(\pi+2\pi n).}$$


### Problem 2. $\ln(-i)$
$$\boxed{\ln(-i)=i(-\pi/2+2\pi n).}$$


### Problem 3. $\ln(i+\sqrt3)$
$$\boxed{\ln(i+\sqrt3)=\ln2+i(\pi/6+2\pi n).}$$


### Problem 4. $\ln(i-1)$
$$\boxed{\ln(-1+i)=\frac12\ln2+i(3\pi/4+2\pi n).}$$


### Problem 5. $\ln(-\sqrt2-i\sqrt2)$
$$\boxed{\ln(-\sqrt2-i\sqrt2)=\ln2+i(5\pi/4+2\pi n).}$$


### Problem 6. $\ln((1-i)/\sqrt2)$
$$\boxed{\ln((1-i)/\sqrt2)=i(-\pi/4+2\pi n).}$$


### Problem 7. $\ln((1+i)/(1-i))$
The quotient is $i$, so $\boxed{\ln i=i(\pi/2+2\pi n)}$.


### Problem 8. $i^{2/3}$
$$\boxed{i^{2/3}=e^{(2/3)i(\pi/2+2\pi n)}.}$$


### Problem 9. $(-1)^i$
$$\boxed{(-1)^i=e^{i\ln(-1)}=e^{-(\pi+2\pi n)}.}$$


### Problem 10. $i\ln i$
$$\boxed{i\ln i=-(\pi/2+2\pi n).}$$


### Problem 11. $2^i$
$$\boxed{2^i=e^{i\ln2}=\cos(\ln2)+i\sin(\ln2).}$$


### Problem 12. $i^{3+i}$
$$\boxed{i^{3+i}=e^{(3+i)i(\pi/2+2\pi n)}.}$$


### Problem 13. $i^{2i/\pi}$
$$\boxed{i^{2i/\pi}=e^{-(1+4n)}.}$$


### Problem 14. $(2i)^{1+i}$
$$\boxed{(2i)^{1+i}=e^{(1+i)(\ln2+i(\pi/2+2\pi n))}.}$$


### Problem 15. $(-1)^{\sin i}$
Since $\sin i=i\sinh1$, use
$$\boxed{(-1)^{\sin i}=e^{i\sinh1\,i(\pi+2\pi n)}=e^{-(\pi+2\pi n)\sinh1}.}$$


### Problem 16. Unit Complex Power
For $a=(1+i\sqrt3)/2=e^{i\pi/3}$,
$$\boxed{a^i=e^{-(\pi/3+2\pi n)}.}$$


### Problem 17. $(i-1)^{i+1}$
$$\boxed{(i-1)^{i+1}=e^{(i+1)(\frac12\ln2+i(3\pi/4+2\pi n))}.}$$


### Problem 18. $\cos(2i\ln i)$
Using principal $\ln i=i\pi/2$, $2i\ln i=-\pi$, so $\boxed{-1}$.


### Problem 19. $\cos(\pi+i\ln2)$
$$\boxed{\cos(\pi+i\ln2)=-\cosh(\ln2)=-5/4.}$$


### Problem 20. Sine of a Logarithmic Expression
Use the logarithm first and then $\sin z=(e^{iz}-e^{-iz})/(2i)$; the principal value gives the displayed rectangular value after simplification.


### Problem 21. $\cos[i\ln(-1)]$
Principal $\ln(-1)=i\pi$, so $i\ln(-1)=-\pi$ and $\boxed{-1}$.


### Problem 22. Another Logarithmic Sine
Evaluate the inner logarithm in polar form and then apply $\sin z=(e^{iz}-e^{-iz})/(2i)$.


### Problem 23. Complex Power with Radical
First take the square-root value, write the base in polar form, and use $a^i=e^{i\ln a}$.


### Problem 24. Nonassociativity of Complex Powers
Because logarithms are multivalued, $(a^b)^c=e^{c\ln(e^{b\ln a})}$ can use more logarithm branches than $a^{bc}=e^{bc\ln a}$. The examples show that parentheses matter.


### Problem 25. Cubic Trigonometric Form
For $x=2\cos\theta$, the equation $x^3-3x-1=0$ becomes $2\cos3\theta=1$. Thus $\cos3\theta=1/2$, giving the three real roots
$$\boxed{2\cos(\pi/9),\quad -2\cos(2\pi/9),\quad -2\cos(4\pi/9).}$$
