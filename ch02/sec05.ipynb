## B. Complex Conjugates


### Problem 25. Conjugates of Algebraic Combinations
For products,
$$
\overline{z_1z_2}=\overline{r_1r_2e^{i(\theta_1+\theta_2)}}=r_1r_2e^{-i(\theta_1+\theta_2)}=\bar z_1\bar z_2.
$$
For quotients the same polar calculation gives $\overline{z_1/z_2}=\bar z_1/\bar z_2$. For differences, rectangular form gives
$$\overline{(x_1+iy_1)-(x_2+iy_2)}=(x_1-x_2)-i(y_1-y_2)=\bar z_1-\bar z_2.$$

## D. Complex Equations


### Problem 35. Equating Real and Imaginary Parts
$$\boxed{x=-4,\quad y=3.}$$


### Problem 36. Linear Equation
From $3+2ix=y-i$, $\boxed{x=-1/2,
\quad y=3}$.


### Problem 37. Zero Complex Number
$$\boxed{x=0,
\quad y=0.}$$


### Problem 38. Another Linear Equation
$$\boxed{x=-7,
\quad y=2.}$$


### Problem 39. Symmetric Equality
$x+iy=y+ix$ implies $\boxed{x=y}$, with any real value allowed.


### Problem 40. Linear Equation with $i x$
Equating real and imaginary parts gives $\boxed{x=0,
\quad y=3}$.


### Problem 41. Two Real Equations
Solve $2x-3y-5=0$ and $x+2y+1=0$:
$$\boxed{x=1,
\quad y=-1.}$$


### Problem 42. Another Pair of Real Equations
Solving $x+2y+3=0$ and $3x-y-1=0$ gives
$$\boxed{x=-1/7,
\quad y=-10/7.}$$


### Problem 43. Quadratic Equation
From $(x+iy)^2=2ix$, the solutions are
$$\boxed{(x,y)=(0,0),(1,1),(-1,1).}$$


### Problem 44. Square of $1-i$
$$(1-i)^2=-2i,$$ so $\boxed{x=0,
\quad y=-2}$.


### Problem 45. Equality of Conjugate Squares
$(x+iy)^2=(x-iy)^2$ gives $4ixy=0$, hence
$$\boxed{x=0\text{ or }y=0.}$$


### Problem 46. Quotient Equation
The equation gives $x=-y$, with $z\ne0$. Thus
$$\boxed{y=-x,
\quad (x,y)\ne(0,0).}$$


### Problem 47. Cube Roots of $-1$
$$\boxed{x+iy\in\left\{-1,\frac12+\frac{\sqrt3}{2}i,\frac12-\frac{\sqrt3}{2}i\right\}.}$$


### Problem 48. Rational Equation
Solving $z+2+3i=(2+i)(2z-3)$ gives
$$\boxed{z=\frac{3-6i}{5},
\quad x=3/5,
\quad y=-6/5.}$$


### Problem 49. Modulus Equal to a Complex Number
The right side must be real, so $y=0$. Then $|1-x|=x$, giving
$$\boxed{x=1/2,
\quad y=0.}$$


### Problem 50. Modulus Equal to $y-ix$
The imaginary part forces $x=0$, and then $|y|=y$. Thus
$$\boxed{x=0,
\quad y\ge0.}$$


## E. Graphs in the Complex Plane


### Problem 51. Circle Centered at the Origin
$$\boxed{|z|=2\text{ is the circle }x^2+y^2=4.}$$


### Problem 52. Imaginary Axis
$$\boxed{\operatorname{Re}z=0\text{ is the imaginary axis.}}$$


### Problem 53. Unit Circle Centered at $1$
$$\boxed{|z-1|=1\text{ is }(x-1)^2+y^2=1.}$$


### Problem 54. Open Disk Centered at $1$
$$\boxed{|z-1|<1\text{ is the open disk }(x-1)^2+y^2<1.}$$


### Problem 55. Horizontal Line
$z-\bar z=2iy=5i$, so $\boxed{y=5/2}$.


### Problem 56. Positive Imaginary Ray
$$\boxed{\arg z=\pi/2\text{ is the positive imaginary axis, excluding }0.}$$


### Problem 57. Hyperbola
$\operatorname{Re}(z^2)=x^2-y^2$, so $\boxed{x^2-y^2=4}$.


### Problem 58. Half-Plane
$$\boxed{x>2.}$$


### Problem 59. Circle Centered at $-3i$
$$\boxed{|z+3i|=4\text{ is }x^2+(y+3)^2=16.}$$


### Problem 60. Circle Centered at $1-i$
$$\boxed{|z-1+i|=2\text{ is }(x-1)^2+(y+1)^2=4.}$$


### Problem 61. Lower Half-Plane
$$\boxed{\operatorname{Im}z<0\text{ means }y<0.}$$


### Problem 62. Ellipse
The sum of distances to $-1$ and $1$ is $8$, so the graph is an ellipse with foci $\pm1$ and major axis $8$.


### Problem 63. Axes
$z^2=\bar z^2$ gives $4ixy=0$, so $\boxed{x=0\text{ or }y=0}$.


### Problem 64. Pair of Diagonal Lines
$z^2=-\bar z^2$ gives $2(x^2-y^2)=0$, so $\boxed{y=\pm x}$.


### Problem 65. Distance Formula
If $z_j=x_j+iy_j$, then
$$|z_1-z_2|=\sqrt{(x_1-x_2)^2+(y_1-y_2)^2}.$$
Thus Problems 53, 54, 59, and 60 are circles or disks centered at the corresponding subtracted point.


## F. Physical Applications


### Problem 66. Position, Velocity, and Acceleration in the Example
For $z=(i+2t)/(t-i)$,
$$z=\frac{(i+2t)(t+i)}{t^2+1}=\frac{2t^2-1}{t^2+1}+i\frac{3t}{t^2+1}.$$
Thus
$$\boxed{x=\frac{2t^2-1}{t^2+1},\quad y=\frac{3t}{t^2+1}.}$$
The complex derivative method gives $v=3/(t^2+1)$ and $a=6/(t^2+1)^{3/2}$, matching the component calculation.


### Problem 67. Speed and Acceleration for a Rational Path
For $z=(1-it)/(2t+i)$,
$$z'=-\frac{1}{(2t+i)^2},\qquad z''=\frac{4}{(2t+i)^3}.$$
Therefore
$$\boxed{v=\frac{1}{4t^2+1},\qquad a=\frac{4}{(4t^2+1)^{3/2}}.}$$


### Problem 68. Uniform Circular Motion
$z=\cos2t+i\sin2t=e^{2it}$. Hence the particle moves on the unit circle with
$$\boxed{v=2,
\quad a=4.}$$
