# 2-12. Hyperbolic Functions


### Problem 1. Formula for $\sin(x+iy)$
Using $\sin z=(e^{iz}-e^{-iz})/(2i)$ gives
$$\boxed{\sin(x+iy)=\sin x\cosh y+i\cos x\sinh y.}$$


### Problem 2. Formula for $\cos(x+iy)$
$$\boxed{\cos(x+iy)=\cos x\cosh y-i\sin x\sinh y.}$$


### Problem 3. Formula for $\sinh(x+iy)$
$$\boxed{\sinh(x+iy)=\sinh x\cos y+i\cosh x\sin y.}$$


### Problem 4. Formula for $\cosh(x+iy)$
$$\boxed{\cosh(x+iy)=\cosh x\cos y+i\sinh x\sin y.}$$


### Problem 5. Double-Angle Sine
The exponential definitions give $\boxed{\sin2z=2\sin z\cos z}$.


### Problem 6. Double-Angle Cosine
$$\boxed{\cos2z=\cos^2z-\sin^2z.}$$


### Problem 7. Double-Angle Hyperbolic Sine
$$\boxed{\sinh2z=2\sinh z\cosh z.}$$


### Problem 8. Double-Angle Hyperbolic Cosine
$$\boxed{\cosh2z=\cosh^2z+\sinh^2z.}$$


### Problem 9. Derivative of Cosine
Termwise or exponential differentiation gives $\boxed{d(\cos z)/dz=-\sin z}$.


### Problem 10. Derivative of Hyperbolic Cosine
$$\boxed{d(\cosh z)/dz=\sinh z.}$$


### Problem 11. Fundamental Hyperbolic Identity
From the definitions,
$$\boxed{\cosh^2z-\sinh^2z=1.}$$


### Problem 12. Fourth-Power Identity
Use $\cos^2z+\sin^2z=1$ and $2\sin z\cos z=\sin2z$:
$$\boxed{\cos^4z+\sin^4z=1-\frac12\sin^22z.}$$


### Problem 13. Triple-Angle Cosine
De Moivre's theorem gives $\boxed{\cos3z=4\cos^3z-3\cos z}$.


### Problem 14. Sine of $iz$
$$\boxed{\sin(iz)=i\sinh z.}$$


### Problem 15. Hyperbolic Sine of $iz$
$$\boxed{\sinh(iz)=i\sin z.}$$


### Problem 16. Tangent of $iz$
$$\boxed{\tan(iz)=i\tanh z.}$$


### Problem 17. Hyperbolic Tangent of $iz$
$$\boxed{\tanh(iz)=i\tan z.}$$


### Problem 18. Tangent of $x+iy$
Dividing the formulas for sine and cosine by $\cos x\cosh y$ gives
$$\boxed{\tan(x+iy)=\frac{\tan x+i\tanh y}{1-i\tan x\tanh y}.}$$


### Problem 19. Hyperbolic Tangent of $x+iy$
$$\boxed{\tanh(x+iy)=\frac{\tanh x+i\tan y}{1+i\tanh x\tan y}.}$$


### Problem 20. Hyperbolic De Moivre Formula
Since $\cosh z+\sinh z=e^z$, raising to the $n$th power gives $e^{nz}=\cosh nz+\sinh nz$. Combining this with $e^{-nz}$ gives the triple-angle identities.


### Problem 21. Graphing Hyperbolic Functions
The plots show $\sinh x$ odd, $\cosh x$ even with minimum $1$, and $\tanh x$ increasing with horizontal asymptotes $\pm1$.


### Problem 22. Power Series
From the exponential definitions,
$$\boxed{\sinh x=\sum_{n=0}^\infty\frac{x^{2n+1}}{(2n+1)!},
\qquad \cosh x=\sum_{n=0}^\infty\frac{x^{2n}}{(2n)!}.}$$


### Problem 23. $\cosh(ix)$
$$\boxed{\cosh(ix)=\cos x,\quad \operatorname{Im}=0,
\quad |\cosh(ix)|=|\cos x|.}$$


### Problem 24. $\cos(ix)$
$$\boxed{\cos(ix)=\cosh x,\quad |\cos(ix)|=\cosh x.}$$


### Problem 25. $\sin(x-iy)$
$$\boxed{\sin(x-iy)=\sin x\cosh y-i\cos x\sinh y.}$$
The absolute value is the square root of the sum of the squares of the displayed real and imaginary parts.


### Problem 26. $\cosh(2-3i)$
$$\boxed{\cosh(2-3i)=\cosh2\cos3-i\sinh2\sin3.}$$


### Problem 27. $\sin(4+3i)$
$$\boxed{\sin(4+3i)=\sin4\cosh3+i\cos4\sinh3.}$$


### Problem 28. $\tanh(1-i\pi)$
Because $\tanh$ has period $i\pi$, $\boxed{\tanh(1-i\pi)=\tanh1}$.


### Problem 29. $\cosh(2\pi i)$
$$\boxed{\cosh(2\pi i)=1.}$$


### Problem 30. $\tanh(3\pi i/4)$
$$\boxed{\tanh(3\pi i/4)=i\tan(3\pi/4)=-i.}$$


### Problem 31. Shifted Hyperbolic Sine
$$\boxed{\sinh(\ln2+i\pi/3)=\frac38+\frac{5\sqrt3}{8}i.}$$


### Problem 32. Shifted Hyperbolic Cosine
$$\boxed{\cosh(i\pi/2-\ln3)=-\frac43 i.}$$


### Problem 33. $\tan i$
$$\boxed{\tan i=i\tanh1.}$$


### Problem 34. $\sin(i\pi/2)$
$$\boxed{\sin(i\pi/2)=i\sinh(\pi/2).}$$


### Problem 35. $\cosh(i\pi+2)$
$$\boxed{\cosh(i\pi+2)=-\cosh2.}$$


### Problem 36. $\sinh(1+i\pi/2)$
$$\boxed{\sinh(1+i\pi/2)=i\cosh1.}$$


### Problem 37. $\cos(i\pi)$
$$\boxed{\cos(i\pi)=\cosh\pi.}$$


### Problem 38. Circular and Hyperbolic Names
If $x=\cos t$, $y=\sin t$, then $x^2+y^2=1$. If $x=\cosh t$, $y=\sinh t$, then $x^2-y^2=1$. This explains the terminology.
