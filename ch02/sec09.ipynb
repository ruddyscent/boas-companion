### Problem 27. Absolute Value of an Imaginary Exponential
For real $y$, $e^{iy}=\cos y+i\sin y$, so $|e^{iy}|=1$. If $z=x+iy$, then $e^z=e^xe^{iy}$ and $\boxed{|e^z|=e^x}$.


### Problem 28. Absolute Value of Products and Quotients
If $z_j=r_je^{i\theta_j}$, then $|z_1z_2|=r_1r_2=|z_1||z_2|$ and $|z_1/z_2|=r_1/r_2=|z_1|/|z_2|$.
