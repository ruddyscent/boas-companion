# 2-11. The Exponential and Trigonometric Functions


### Problem 1. Series Definitions
Comparing the series for $e^{iz}$ with the even and odd power series gives
$$e^{iz}=\cos z+i\sin z,$$
so
$$\boxed{\sin z=\frac{e^{iz}-e^{-iz}}{2i},
\qquad \cos z=\frac{e^{iz}+e^{-iz}}2.}$$


### Problem 2. Solving Euler's Pair
Adding and subtracting $e^{i\theta}=\cos\theta+i\sin\theta$ and $e^{-i\theta}=\cos\theta-i\sin\theta$ gives
$$\boxed{\cos\theta=\frac{e^{i\theta}+e^{-i\theta}}2,
\qquad \sin\theta=\frac{e^{i\theta}-e^{-i\theta}}{2i}.}$$


### Problem 3. Exponential Value
$$\boxed{e^{-i\pi/4+\ln3}=\frac{3\sqrt2}{2}-\frac{3\sqrt2}{2}i.}$$


### Problem 4. Real Exponential Times $e^{-i\pi}$
$$\boxed{e^{3\ln2-i\pi}=-8.}$$


### Problem 5. Mixed Exponent
$$\boxed{e^{i\pi/4+(\ln2)/2}=1+i.}$$


### Problem 6. Cosine of a Pure Imaginary Number
$$\boxed{\cos(i\ln5)=\cosh(\ln5)=13/5.}$$


### Problem 7. Tangent of a Pure Imaginary Number
$$\boxed{\tan(i\ln2)=i\tanh(\ln2)=\frac35i.}$$


### Problem 8. Shifted Cosine
$$\boxed{\cos(\pi-2i\ln3)=-\cosh(2\ln3)=-41/9.}$$


### Problem 9. Shifted Sine
$$\boxed{\sin(\pi-i\ln3)=\frac43i.}$$


### Problem 10. Nested Logarithm
Using the principal value $\ln i=i\pi/2$,
$$\boxed{\sin(i\ln i)=\sin(-\pi/2)=-1.}$$


### Problem 11. Orthogonality of Cosines
Writing cosines as exponentials, every exponential term has nonzero integer frequency over $[-\pi,\pi]$. Hence the integral is $\boxed{0}$.


### Problem 12. Square of a Cosine
$\cos^2 3x=(1+\cos6x)/2$, so
$$\boxed{\int_{-\pi}^{\pi}\cos^2 3x\,dx=\pi.}$$


### Problem 13. Orthogonality of Sines
The product is a sum of cosines with nonzero integer frequencies, so the integral is $\boxed{0}$.


### Problem 14. Square of a Sine
$\sin^2 4x=(1-\cos8x)/2$, so
$$\boxed{\int_0^{2\pi}\sin^2 4x\,dx=\pi.}$$


### Problem 15. Sine-Cosine Orthogonality
The product is a sum of sine terms over a symmetric interval, so the integral is $\boxed{0}$.


### Problem 16. Another Sine-Cosine Orthogonality Integral
Again the exponential terms have nonzero integer frequencies over $[-\pi,\pi]$, so the integral is $\boxed{0}$.


### Problem 17. Integral of $e^{ax}\cos bx$
Since $\int e^{(a+ib)x}dx=e^{(a+ib)x}/(a+ib)$, taking real parts gives
$$\boxed{\int e^{ax}\cos bx\,dx=\frac{e^{ax}(a\cos bx+b\sin bx)}{a^2+b^2}.}$$


### Problem 18. Integral of $e^{ax}\sin bx$
Taking imaginary parts gives
$$\boxed{\int e^{ax}\sin bx\,dx=\frac{e^{ax}(a\sin bx-b\cos bx)}{a^2+b^2}.}$$
