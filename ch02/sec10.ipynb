### Problem 27. Double-Angle Formulas
From $(\cos\theta+i\sin\theta)^2=\cos2\theta+i\sin2\theta$,
$$\boxed{\cos2\theta=\cos^2\theta-\sin^2\theta,
\qquad \sin2\theta=2\sin\theta\cos\theta.}$$


### Problem 28. Triple-Angle Formulas
From $(\cos\theta+i\sin\theta)^3=\cos3\theta+i\sin3\theta$,
$$\boxed{\cos3\theta=4\cos^3\theta-3\cos\theta,}$$
$$\boxed{\sin3\theta=3\sin\theta-4\sin^3\theta.}$$


### Problem 29. Center of Mass
Identifying $z_j=x_j+iy_j$, the coordinate average is
$$\boxed{\frac{z_1+z_2+z_3}{3}=\frac{x_1+x_2+x_3}{3}+i\frac{y_1+y_2+y_3}{3}.}$$


### Problem 30. Sum of Cube Roots of $8$
The roots are $2,2\omega,2\omega^2$ where $1+\omega+\omega^2=0$. Thus their sum is $0$.


### Problem 31. Sum of $n$th Roots
The roots are $\alpha\omega^k$, $k=0,\ldots,n-1$, with $\omega=e^{2\pi i/n}$. Since $1+\omega+\cdots+\omega^{n-1}=0$, their sum is $0$.


### Problem 32. Cube Roots of Unity
The cube roots are $1$, $\omega=e^{2\pi i/3}$, and $\omega^2=e^{4\pi i/3}$. Also $\omega^3=1$ and $\omega^2=\bar\omega$, so the notation $1,\omega,\omega^2$ is consistent.
