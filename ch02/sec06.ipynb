# 2-6. Complex Infinite Series


### Problem 1. Absolute Convergence Implies Convergence
Let the terms be $a_n+ib_n$. Since $|a_n|\le\sqrt{a_n^2+b_n^2}$ and $|b_n|\le\sqrt{a_n^2+b_n^2}$, convergence of $\sum\sqrt{a_n^2+b_n^2}$ implies convergence of $\sum a_n$ and $\sum b_n$ by comparison. Hence $\sum(a_n+ib_n)$ converges.


### Problem 2. Powers of $1+i$
The term magnitudes are $(\sqrt2)^n$, so the terms do not approach zero. The series diverges.


### Problem 3. Reciprocal Powers
This is geometric with ratio $1/(1+i)$, whose modulus is $1/\sqrt2<1$. It converges absolutely.


### Problem 4. Unit-Modulus Ratio
$(1-i)/(1+i)=-i$, so the terms do not tend to zero. The series diverges.


### Problem 5. Real and Imaginary Parts
The real part $\sum 1/n^2$ converges, but the imaginary part $\sum 1/n$ diverges. The complex series diverges.


### Problem 6. Constant Complex Multiple
$\sum(1+i)/n^2$ converges absolutely because $\sum |1+i|/n^2=\sqrt2\sum1/n^2$ converges.


### Problem 7. Exponentially Growing Terms
$|(i-1)^n/n|=(\sqrt2)^n/n$, so the terms do not approach zero. The series diverges.


### Problem 8. Roots of Unity Terms
$|e^{in\pi/6}|=1$, so the terms do not approach zero. The series diverges.


### Problem 9. Alternating Residue Classes
The series $\sum i^n/n$ is not absolutely convergent, but its real and imaginary parts are alternating harmonic subseries. It converges conditionally.


### Problem 10. Geometric Series
The ratio has modulus
$$\left|\frac{1+i}{1-i\sqrt3}\right|=\frac{\sqrt2}{2}<1,$$
so the series converges absolutely.


### Problem 11. Squared Geometric Ratio
The ratio modulus is
$$\left|\frac{2+i}{3-4i}\right|^2=\left(\frac{\sqrt5}{5}\right)^2=\frac15<1.$$
The series converges absolutely.


### Problem 12. Factorial Denominator
The ratio test gives $|3+2i|/(n+1)\to0$, so the series converges absolutely.


### Problem 13. Geometric Series
$$\left|\frac{1+i}{2-i}\right|=\sqrt{2/5}<1,$$
so the series converges absolutely.


### Problem 14. Ratio Test Divergence
If the ratio-test limit $\rho>1$, then eventually $|a_{n+1}|>|a_n|$ by a fixed factor greater than $1$ after passing to a subsequence bound. In particular the terms cannot tend to zero. Therefore the complex series diverges.
