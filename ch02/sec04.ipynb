# 2-4. Terminology and Notation

For each number, write $z=x+iy=re^{i\theta}$ and $\bar z=x-iy=re^{-i\theta}$.


### Problem 1. First-Quadrant Unit Point
$z=1+i$, so $x=1$, $y=1$, $r=\sqrt2$, $\theta=\pi/4$.
$$\boxed{z=1+i=\sqrt2(\cos\pi/4+i\sin\pi/4)=\sqrt2e^{i\pi/4},\quad \bar z=1-i.}$$


### Problem 2. Second-Quadrant Unit Point
$z=i-1=-1+i$, so $r=\sqrt2$, $\theta=3\pi/4$.
$$\boxed{z=-1+i=\sqrt2e^{3\pi i/4},\quad \bar z=-1-i.}$$


### Problem 3. Fourth-Quadrant Point
$z=1-i\sqrt3$, so $r=2$, $\theta=-\pi/3$.
$$\boxed{z=2e^{-i\pi/3},\quad \bar z=1+i\sqrt3=2e^{i\pi/3}.}$$


### Problem 4. Second-Quadrant Point
$z=-\sqrt3+i$, so $r=2$, $\theta=5\pi/6$.
$$\boxed{z=2e^{5\pi i/6},\quad \bar z=-\sqrt3-i=2e^{-5\pi i/6}.}$$


### Problem 5. Positive Imaginary Axis
$$\boxed{2i=2e^{i\pi/2},\quad \overline{2i}=-2i.}$$


### Problem 6. Negative Imaginary Axis
$$\boxed{-4i=4e^{-i\pi/2},\quad \overline{-4i}=4i.}$$


### Problem 7. Negative Real Axis
$$\boxed{-1=e^{i\pi},\quad \bar z=-1.}$$


### Problem 8. Positive Real Axis
$$\boxed{3=3e^{0i},\quad \bar z=3.}$$


### Problem 9. Second-Quadrant Point
$z=-2+2i$, so $r=2\sqrt2$, $\theta=3\pi/4$.
$$\boxed{z=2\sqrt2e^{3\pi i/4},\quad \bar z=-2-2i.}$$


### Problem 10. Fourth-Quadrant Point
$z=2-2i$, so $r=2\sqrt2$, $\theta=-\pi/4$.
$$\boxed{z=2\sqrt2e^{-i\pi/4},\quad \bar z=2+2i.}$$


### Problem 11. Polar Form to Rectangular Form
$$2(\cos\pi/6+i\sin\pi/6)=\boxed{\sqrt3+i},\qquad \bar z=\sqrt3-i.$$


### Problem 12. Negative Sine Sign
$$4(\cos 2\pi/3-i\sin2\pi/3)=4e^{-2\pi i/3}=\boxed{-2-2\sqrt3 i}.$$
The conjugate is $-2+2\sqrt3 i$.


### Problem 13. Unit Negative Imaginary Point
$$\cos(3\pi/2)+i\sin(3\pi/2)=\boxed{-i},\qquad \bar z=i.$$


### Problem 14. Radius Two at $45^\circ$
$$2(\cos\pi/4+i\sin\pi/4)=\boxed{\sqrt2+i\sqrt2},\qquad \bar z=\sqrt2-i\sqrt2.$$


### Problem 15. Negative Unit Point
$$\cos\pi-i\sin\pi=\boxed{-1},\qquad \bar z=-1.$$


### Problem 16. Positive Real Radius Five
$$\boxed{5(\cos0+i\sin0)=5},\qquad \bar z=5.$$


### Problem 17. Exponential Form
$$\sqrt2e^{-i\pi/4}=\sqrt2(\cos(-\pi/4)+i\sin(-\pi/4))=\boxed{1-i}.$$
The conjugate is $1+i$.


### Problem 18. Positive Imaginary Radius Three
$$\boxed{3e^{i\pi/2}=3i},\qquad \bar z=-3i.$$


### Problem 19. Degree Measure
$$\boxed{5(\cos20^\circ+i\sin20^\circ)=5e^{20^\circ i}},\qquad \bar z=5e^{-20^\circ i}.$$


### Problem 20. Negative Angle in Degree Form
$$7(\cos110^\circ-i\sin110^\circ)=\boxed{7e^{-110^\circ i}}.$$
Its conjugate is $7e^{110^\circ i}$.
