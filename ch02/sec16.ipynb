# 2-16. Some Applications


### Problem 1. Multiplication by $i$ and Uniform Circular Motion
If $z=x+iy$, then $iz=-y+ix$, which is the point rotated $90^\circ$ counterclockwise. For $z=ae^{i\omega t}$,
$$z'=i\omega ae^{i\omega t},\qquad z''=-\omega^2ae^{i\omega t}.$$
Thus $\boxed{v=a\omega}$ and $\boxed{|a_{\text{cent}}|=a\omega^2=v^2/a}$ toward the center.


### Problem 2. Circle of Radius Five
For $z=5e^{i\omega t}$,
$$\boxed{v=5|\omega|,
\quad |a|=5\omega^2.}$$
The motion is uniform circular motion centered at the origin.


### Problem 3. Shifted Radius
$z=(1+i)e^{it}$ has radius $|1+i|=\sqrt2$, so $\boxed{v=\sqrt2,
|a|=\sqrt2}$ on a circle centered at the origin.


### Problem 4. Straight-Line Motion
$z=(1+i)t-(2+i)(1-t)=(3+2i)t-(2+i)$. Hence the particle moves along a straight line with constant speed
$$\boxed{|z'|=|3+2i|=\sqrt{13}},
\quad \boxed{|z''|=0.}$$


### Problem 5. General Straight-Line Interpolation
$z=z_1t+z_2(1-t)=z_2+(z_1-z_2)t$, so motion is along the line through $z_1,z_2$ with speed $\boxed{|z_1-z_2|}$ and zero acceleration.


### Problem 6. Series and Parallel Impedances
(a) Series: $\boxed{3-2i}$. Parallel: $[(2+3i)^{-1}+(1-5i)^{-1}]^{-1}$.

(b) Convert to rectangular or use polar form; series is $Z_1+Z_2$, parallel is $Z_1Z_2/(Z_1+Z_2)$.


### Problem 7. More Impedance Combinations
Use the same formulas:
$$\boxed{Z_{\rm series}=Z_1+Z_2,
\qquad Z_{\rm parallel}=\frac{Z_1Z_2}{Z_1+Z_2}.}$$
Substitute the rectangular or polar data given in each part.


### Problem 8. R-L Series in Parallel with C
The series branch has $Z_{RL}=R+i\omega L$ and the capacitor has $Z_C=1/(i\omega C)$. Thus
$$\boxed{Z=\frac{Z_{RL}Z_C}{Z_{RL}+Z_C}.}$$
Resonance occurs when the imaginary part of $Z$ is zero.


### Problem 9. Series RLC Circuit
For $Z=R+i(\omega L-1/\omega C)$, angle $45^\circ$ means imaginary part equals $R$:
$$\boxed{\omega L-\frac1{\omega C}=R.}$$
Resonance means $\omega L=1/(\omega C)$, so $\boxed{\omega=1/\sqrt{LC}}$.


### Problem 10. Parallel RLC Circuit
For parallel branches,
$$\boxed{\frac1Z=\frac1R+\frac1{i\omega L}+i\omega C.}$$
Use the real and imaginary parts of $1/Z$ for the angle condition. Resonance occurs when $\omega C-1/(\omega L)=0$, hence $\boxed{\omega=1/\sqrt{LC}}$.


### Problem 11. Sums of Odd Sines and Cosines
Sum $\sum_{k=1}^n e^{i(2k-1)\theta}$ as a geometric series. Taking real and imaginary parts gives the displayed sine-over-sine formulas.


### Problem 12. Optical Intensity Series
The two real sums are the real and imaginary parts of
$$\sum_{n=0}^\infty r^{2n}e^{in\theta}=\frac1{1-r^2e^{i\theta}}.$$
Thus the intensity expression is
$$\boxed{\left|\frac1{1-r^2e^{i\theta}}\right|^2=\frac1{1-2r^2\cos\theta+r^4}.}$$


### Problem 13. Simple Harmonic Motion Solutions
For $y=e^{\pm i\omega t}$, $y''=-\omega^2y$. For $\sin\omega t$ and $\cos\omega t$, the same identity follows by ordinary differentiation. All satisfy $y''=-\omega^2y$.
