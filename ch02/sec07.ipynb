# 2-7. Complex Power Series; Disk of Convergence


### Problem 1. Exponential Series
The ratio test gives $|z|/(n+1)\to0$, so the disk is the whole complex plane.


### Problem 2. Logarithmic-Type Series
The ratio test gives $|z|$. Thus the radius is $1$ and the disk is $\boxed{|z|<1}$.


### Problem 3. Sine-Quotient Type Series
The factorials force ratio limit $0$, so the series converges for every $z$.


### Problem 4. Geometric Series
$\sum z^n$ converges for $\boxed{|z|<1}$.


### Problem 5. Scaled Geometric Series
$\sum(z/2)^n$ converges for $\boxed{|z|<2}$.


### Problem 6. Polynomial Factor with Geometric Part
The $n^2$ factor does not change the radius. Since the geometric factor is $(3iz)^n$, the disk is $\boxed{|z|<1/3}$.


### Problem 7. Cosine-Type Series
The factorial denominator gives convergence for all $z$.


### Problem 8. Factorial Denominator with Even Powers
The ratio test gives limit $0$, so the series converges for all $z$.


### Problem 9. Square-Root Denominator
The ratio test gives $|z|$, so the disk is $\boxed{|z|<1}$.


### Problem 10. Power over $n^2$
The ratio test gives $|z|$, so the disk is $\boxed{|z|<1}$.


### Problem 11. Cubic Factorial Ratio
For $a_n=(n!)^3/(3n)!$, $|a_{n+1}/a_n|\to1/27$. Thus the disk is $\boxed{|z|<27}$.


### Problem 12. Double Factorial Ratio
For $a_n=(n!)^2/(2n)!$, $|a_{n+1}/a_n|\to1/4$. Thus the disk is $\boxed{|z|<4}$.


### Problem 13. Centered Logarithmic Series
The ratio test gives $|z-i|$, so the disk is $\boxed{|z-i|<1}$.


### Problem 14. Polynomial Coefficients Centered at $2i$
The coefficient $n(n+1)$ does not change the radius. The disk is $\boxed{|z-2i|<1}$.


### Problem 15. Centered Geometric Series
The terms are $[(z-2+i)/2]^n$, so the disk is $\boxed{|z-(2-i)|<2}$.


### Problem 16. Even Powers
The series is geometric in $2(z+i-3)^2$. Hence
$$\boxed{|z-(3-i)|<1/\sqrt2.}$$


### Problem 17. Series Verification
Using
$$\frac{\sin z}{z}=\sum_{k=0}^\infty\frac{(-1)^kz^{2k}}{(2k+1)!},
\qquad \frac1{1+z^2}=\sum_{j=0}^\infty(-1)^jz^{2j},$$
Cauchy multiplication gives
$$\boxed{\frac{\sin z}{z(1+z^2)}=\sum_{n=0}^\infty(-1)^nz^{2n}\sum_{k=0}^n\frac1{(2k+1)!}.}$$
The inner sum tends to $\sinh1$, so the ratio test for the outer even-power series gives convergence for $|z|<1$.
