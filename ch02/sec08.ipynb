# 2-8. Elementary Functions of Complex Numbers


### Problem 1. Product Law for the Exponential
Multiply the two absolutely convergent power series:
$$e^{z_1}e^{z_2}=\sum_{m,n\ge0}\frac{z_1^mz_2^n}{m!n!}=\sum_{k=0}^\infty\frac{(z_1+z_2)^k}{k!}=e^{z_1+z_2}.$$


### Problem 2. Derivative of the Exponential
Term-by-term differentiation gives
$$\frac{d}{dz}e^z=\frac{d}{dz}\sum_{n=0}^\infty\frac{z^n}{n!}=\sum_{n=1}^\infty\frac{nz^{n-1}}{n!}=\sum_{m=0}^\infty\frac{z^m}{m!}=e^z.$$


### Problem 3. Series for $e^x\cos x$ and $e^x\sin x$
From $e^{x+iy}=e^x(\cos y+i\sin y)$ and the series for $e^z$,
$$e^x\cos y=\Re\sum_{n=0}^\infty\frac{(x+iy)^n}{n!},
\qquad e^x\sin y=\Im\sum_{n=0}^\infty\frac{(x+iy)^n}{n!}.$$
Putting $y=x$ gives
$$\boxed{e^x\cos x=1+x-\frac{x^3}{3}-\frac{x^4}{6}+\cdots,}$$
$$\boxed{e^x\sin x=x+x^2+\frac{x^3}{3}-\frac{x^5}{30}+\cdots.}$$
